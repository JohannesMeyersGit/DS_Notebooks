## Maximum-Likelihood-Schätzung (MLE) – Schritt-für-Schritt Erklärung

---

### 1. Likelihood-Funktion

Angenommen, wir haben eine Stichprobe $X_1, X_2, \dots, X_n$ aus einer Verteilung mit unbekanntem Parameter $\theta$.  

Die **Likelihood-Funktion** ist die Wahrscheinlichkeit, genau diese Daten unter $\theta$ zu beobachten:

$$
L(\theta) = P(X_1, X_2, \dots, X_n \mid \theta)
$$

Wenn die $X_i$ **unabhängig** sind, können wir die gemeinsame Wahrscheinlichkeit als Produkt der Einzelwahrscheinlichkeiten schreiben:

$$
L(\theta) = \prod_{i=1}^{n} f(X_i \mid \theta)
$$

- $f(X_i \mid \theta)$ = Wahrscheinlichkeitsfunktion (diskret) oder Wahrscheinlichkeitsdichte (stetig) für den Beobachtungswert $X_i$.

---

### 2. Von Likelihood zu Log-Likelihood

Die Likelihood ist ein Produkt:

$$
L(\theta) = f(X_1 \mid \theta) \cdot f(X_2 \mid \theta) \cdot \dots \cdot f(X_n \mid \theta)
$$

**Problem:** Produkte vieler kleiner Wahrscheinlichkeiten werden sehr klein → numerisch instabil.  

**Lösung:** Logarithmus nehmen.  
- Logarithmus wandelt **Produkte in Summen** um:

$$
\ell(\theta) = \log L(\theta) = \log \prod_{i=1}^{n} f(X_i \mid \theta) = \sum_{i=1}^{n} \log f(X_i \mid \theta)
$$

**Wichtig:** Es passiert **nicht mehr** als mathematisches Umformen: Produkt → Summe.  
- Maximieren von $L(\theta)$ ist **äquivalent** zum Maximieren von $\ell(\theta)$, da $\log$ eine streng monoton wachsende Funktion ist.

---

### 3. Maximierung der Log-Likelihood

Um den MLE zu finden:

1. Schreibe die Log-Likelihood $\ell(\theta)$ auf.
2. Berechne die Ableitung nach $\theta$: $\frac{d\ell}{d\theta}$.
3. Setze die Ableitung auf Null: $\frac{d\ell}{d\theta} = 0$.
4. Löse nach $\theta$ auf: $\hat{\theta}_{\mathrm{MLE}}$.

> Optional kann man die zweite Ableitung überprüfen, ob es tatsächlich ein Maximum ist.

---

### 4. Beispiel: Bernoulli

- Stichprobe: $X_i \in \{0,1\}$, Parameter $p = P(X_i=1)$  
- Likelihood:

$$
L(p) = \prod_{i=1}^{n} p^{X_i} (1-p)^{1-X_i}
$$

- Log-Likelihood:

$$
\ell(p) = \sum_{i=1}^{n} \log \big( p^{X_i} (1-p)^{1-X_i} \big) 
= \sum_{i=1}^{n} X_i \log p + (1-X_i) \log(1-p)
$$

- Ableitung nach $p$:

$$
\frac{d\ell}{dp} = \sum_{i=1}^{n} \frac{X_i}{p} - \frac{1-X_i}{1-p}
$$

- Gleich Null setzen und lösen:

$$
\sum_{i=1}^{n} \frac{X_i}{\hat p} - \frac{1-X_i}{1-\hat p} = 0 
\quad\Rightarrow\quad
\hat p = \frac{1}{n} \sum_{i=1}^{n} X_i
$$

> Ergebnis: MLE = **Stichprobenmittelwert**.

---

### 5. Sufficient Statistics (ausreichende Statistik)

Eine Statistik $T(X_1,\dots,X_n)$ ist **sufficient** für einen Parameter $\theta$, wenn **T alle Informationen enthält, die die Stichprobe über $\theta$ liefert**.  

**Anschauliches Beispiel:**

- Bernoulli: $X_1,\dots,X_n \in \{0,1\}$, Parameter $p$  
- Betrachten wir die Summe $T = \sum X_i$.  
- **Frage:** Brauchen wir die einzelnen 0/1-Werte noch, wenn wir $T$ kennen?  
- **Antwort:** Nein. Wenn wir wissen, wie viele Erfolge es gibt, wissen wir alles, was über $p$ relevant ist.  
- Daher ist $T = \sum X_i$ **sufficient statistic** für $p$.

**Warum wichtig für MLE:**  
- Die Likelihood hängt nur von sufficient statistics ab.  
- Wir müssen also **nicht alle $X_i$ einzeln** berücksichtigen, sondern nur die ausreichende Statistik.  

**Ein anderes Beispiel: Normalverteilung**

- $X_i \sim \mathcal{N}(\mu, \sigma^2)$  
- Sufficient statistics: $T_1 = \sum X_i$, $T_2 = \sum X_i^2$  
- Diese beiden Zahlen enthalten **alle Information über $\mu$ und $\sigma^2$**, die in der Stichprobe steckt.  

> Vorteil: Reduziert die Datenmenge, die wir für die MLE-Berechnung benötigen.

### 5. Sufficient Statistics (ausreichende Statistik)

Eine Statistik $T(X_1, \dots, X_n)$ ist **sufficient** für einen Parameter $\theta$, wenn sie **alle Informationen enthält, die die Stichprobe über $\theta$ liefert**.  
Das bedeutet: **Kennt man $T$, liefert die ursprüngliche Stichprobe keine weiteren Informationen für die Parameterschätzung.**

---

#### Anschauliches Beispiel: Bernoulli-Verteilung

- Stichprobe: $X_1, \dots, X_n \in \{0,1\}$, Parameter $p$ (Erfolgswahrscheinlichkeit)  
- Betrachten wir die Summe der Erfolge:  
  $$
  T = \sum_{i=1}^n X_i
  $$
- **Frage:** Brauchen wir die einzelnen 0/1-Werte noch, wenn wir $T$ kennen?  
- **Antwort:** Nein. Kennt man $T$, wissen wir, wie viele Erfolge es gibt – das ist **alle Information**, die für die Schätzung von $p$ relevant ist.  
- Daher ist $T = \sum X_i$ eine **sufficient statistic** für $p$.

**Grafische Vorstellung:**  
- Jede Stichprobe von $n=10$ Bernoulli-Versuchen → unterschiedliche 0/1-Folgen.  
- Die Summe $T$ reduziert diese Vielfalt auf eine einzelne Zahl.  
- **Alle Sequenzen mit der gleichen Summe T liefern die gleiche Likelihood für p.**

---

#### Warum wichtig für MLE

- Die Likelihoodfunktion hängt **nur von den sufficient statistics ab**, nicht von allen einzelnen Beobachtungen:  
  $$
  L(p; X_1, \dots, X_n) = L(p; T)
  $$
- Vorteil: **Reduziert den Rechenaufwand** erheblich, besonders bei großen Stichproben.  
- Intuition: Für die MLE-Berechnung müssen wir **nicht alle $X_i$ einzeln** berücksichtigen, sondern nur die ausreichende Statistik.

---

#### Beispiel: Normalverteilung

- Stichprobe: $X_1, \dots, X_n \sim \mathcal{N}(\mu, \sigma^2)$  
- Sufficient statistics:  
  $$
  T_1 = \sum X_i \quad \text{(Summe der Werte, enthält Information über μ)}
  $$  
  $$
  T_2 = \sum X_i^2 \quad \text{(Summe der Quadrate, enthält Information über σ²)}
  $$  
- Kennt man $T_1$ und $T_2$, liefert die ursprüngliche Stichprobe keine weiteren Informationen für MLE von $\mu$ und $\sigma^2$.  

**Intuition:**  
- $T_1$ → Mittelwert  
- $T_2$ → Streuung  
- Alles, was man für die Likelihood der Normalverteilung braucht, steckt in diesen beiden Zahlen.  

---

#### Vorteil für Statistik & Data Science

- **Datenreduktion:** Große Stichprobe → kleine sufficient statistics  
- **Effizienz:** MLE-Parameter lassen sich schneller und stabiler berechnen  
- **Robustheit:** Beobachtet man wiederholt die sufficient statistics, stabilisieren sich MLE-Schätzungen → Modell ist robust

> Merksatz: „Kennt man die sufficient statistics, kann man alle einzelnen Datenpunkte ignorieren, ohne Information zu verlieren.“


---

### 6. Zusammenfassung der Schritte MLE

1. Bestimme die Likelihood $L(\theta)$.
2. Schreibe Log-Likelihood $\ell(\theta) = \log L(\theta)$.
3. Berechne die Ableitung $\frac{d\ell}{d\theta}$.
4. Setze Ableitung = 0 und löse nach $\hat{\theta}_{\mathrm{MLE}}$.
5. Optional: Prüfe zweite Ableitung für Maximum.
6. Beachte sufficient statistics – oft reicht eine Zusammenfassung der Daten.

---



```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, IntSlider, FloatSlider

np.random.seed(42)

def bernoulli_mle_animated(n=20, p_true=0.7):
    # Schritt 0: Stichprobe
    data = np.random.binomial(1, p=p_true/100, size=n)
    successes = np.sum(data)
    
    theta_values = np.linspace(0.01, 0.99, 200)
    
    # Likelihood
    L = theta_values**successes * (1 - theta_values)**(n - successes)
    
    # Log-Likelihood
    logL = np.log(L)
    
    # Ableitung
    dlogL = successes/theta_values - (n-successes)/(1-theta_values)
    
    # MLE
    theta_mle = successes / n
    
    fig, axs = plt.subplots(1,3, figsize=(18,5))
    
    # Log-Likelihood
    axs[0].plot(theta_values, logL, color='green')
    axs[0].set_title("Log-Likelihood ℓ(θ)")
    axs[0].set_xlabel("θ")
    axs[0].set_ylabel("ℓ(θ)")
    
    # Ableitung
    axs[1].plot(theta_values, dlogL, color='purple')
    axs[1].axhline(0, color='black', linestyle='--')
    axs[1].set_title("Ableitung dℓ/dθ")
    axs[1].set_xlabel("θ")
    axs[1].set_ylabel("dℓ/dθ")
    
    # Animation MLE-Punkt Schritt für Schritt
    idx = int(theta_mle * len(theta_values))
    axs[0].plot(theta_values[idx], logL[idx], 'ro', label=r'$\hat{\theta}_{MLE}$')
    axs[1].plot(theta_values[idx], dlogL[idx], 'ro', label=r'$\hat{\theta}_{MLE}$')
    axs[0].legend()
    axs[1].legend()
    
    # Sufficient statistic
    axs[2].bar([0,1], [successes, n-successes], color=['blue','gray'])
    axs[2].set_xticks([0,1])
    axs[2].set_xticklabels(['Erfolg (1)','Misserfolg (0)'])
    axs[2].set_ylabel("Anzahl")
    axs[2].set_title(f"Sufficient Statistic: Summe der Erfolge = {successes}")
    
    plt.tight_layout()
    plt.show()
    
    print(f"MLE für θ (Anteil Erfolge): {theta_mle:.2f}")
    print("Sufficient statistic: Summe der Erfolge enthält alle Information über θ, die die Stichprobe liefert.")
    print("Für Normalverteilung: MLE schätzt Mittelwert μ = Stichprobenmittelwert und Varianz σ² = Mittel der quadrierten Abweichungen vom Mittelwert (ddof=0).")

interact(
    bernoulli_mle_animated,
    n=IntSlider(min=5, max=100, step=5, value=20, description="Stichprobengröße"),
    p_true=IntSlider(min=1, max=99, step=1, value=70, description="Wahrer p (%)")
);
```

## Interaktives Bernoulli-MLE Widget – Erklärung

Dieses Widget visualisiert **die Maximum-Likelihood-Schätzung (MLE)** für eine Bernoulli-Verteilung Schritt für Schritt und macht den Begriff der **sufficient statistic** anschaulich.

---

### 1. Stichprobe

- Mit dem Slider "Stichprobengröße" kann man die Anzahl der Beobachtungen $n$ einstellen.  
- Mit dem Slider "Wahrer p (%)" wird die zugrunde liegende Erfolgswahrscheinlichkeit $p$ der Bernoulli-Verteilung festgelegt.  
- Das Widget erzeugt zufällige Stichprobenwerte $X_i \in \{0,1\}$.

---

### 2. Likelihood L(θ)

- Zeigt die Wahrscheinlichkeit, die beobachtete Stichprobe für verschiedene θ-Werte zu erhalten:  
  $$
  L(\theta) = \prod_{i=1}^{n} \theta^{X_i} (1-\theta)^{1-X_i}
  $$
- Höhepunkt der Kurve entspricht den Werten von θ, bei denen die Daten am wahrscheinlichsten sind.

---

### 3. Log-Likelihood ℓ(θ)

- Um numerische Stabilität zu gewährleisten, wird der Logarithmus genommen:  
  $$
  \ell(\theta) = \log L(\theta) = \sum_{i=1}^{n} X_i \log \theta + (1-X_i) \log (1-\theta)
  $$
- Das Maximum der Log-Likelihood ist identisch mit dem Maximum der Likelihood, aber leichter zu visualisieren und zu berechnen.

---

### 4. Ableitung der Log-Likelihood

- Die Ableitung zeigt, wo die Log-Likelihood steigt und fällt:  
  $$
  \frac{d\ell}{d\theta} = \frac{\sum X_i}{\theta} - \frac{n-\sum X_i}{1-\theta}
  $$
- Die Nullstelle der Ableitung entspricht dem **MLE-Punkt**: $\hat{\theta}_{MLE} = \frac{\sum X_i}{n}$.  
- Im Widget ist dieser Punkt **rot markiert**, sowohl auf der Log-Likelihood-Kurve als auch auf der Ableitung.

---

### 5. Sufficient Statistic

- Die **sufficient statistic** ist die Summe der Erfolge: $T = \sum X_i$.  
- Sie enthält **alle Information**, die die Stichprobe über den Parameter θ liefert.  
- Die einzelnen Beobachtungen sind nicht mehr nötig, um MLE zu berechnen.  
- Im Widget wird die Anzahl der Erfolge und Misserfolge als Balkendiagramm dargestellt.

---

### 6. Zusammenfassung

- **MLE** findet den Parameter θ, unter dem die beobachteten Daten am wahrscheinlichsten sind.  
- Für Bernoulli: MLE = **Stichprobenmittelwert der Erfolge**.  
- Für Normalverteilung: MLE schätzt **Mittelwert μ = Stichprobenmittelwert** und **Varianz σ² = Mittel der quadrierten Abweichungen (ddof=0)**.  
- Mit steigender Stichprobengröße nähert sich der MLE dem wahren Wert von θ an und die Likelihood-Kurve wird schmaler.  

> Mit dem interaktiven Widget kann man live beobachten, wie Stichprobengröße, Log-Likelihood, Ableitung und sufficient statistic zusammenwirken, um den MLE zu bestimmen.


```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, IntSlider, FloatSlider

np.random.seed(42)

def normal_mle_animated(n=30, mu_true=5.0, sigma_true=2.0):
    # 1. Stichprobe
    data = np.random.normal(loc=mu_true, scale=sigma_true, size=n)
    x_vals = np.linspace(min(data)-3*sigma_true, max(data)+3*sigma_true, 200)
    
    # 2. MLE Berechnung
    mu_mle = np.mean(data)
    sigma2_mle = np.var(data)
    sigma_mle = np.sqrt(sigma2_mle)
    
    # 3. Log-Likelihood für mu (σ² fix)
    mu_values = np.linspace(mu_true - 3, mu_true + 3, 200)
    logL_mu = [-0.5*np.sum((data - mu_val)**2)/sigma2_mle for mu_val in mu_values]
    
    # 4. Log-Likelihood für sigma (μ fix)
    sigma_values = np.linspace(0.5, sigma_true*3, 200)
    logL_sigma = [-0.5*n*np.log(2*np.pi*sigma_val**2) - np.sum((data-mu_mle)**2)/(2*sigma_val**2)
                  for sigma_val in sigma_values]
    
    # Subplots
    fig, axs = plt.subplots(2,3, figsize=(20,10))
    
    # A: Histogram + Normalverteilungen
    axs[0,0].hist(data, bins='auto', density=True, alpha=0.6, color='gray', label='Stichprobe')
    axs[0,0].plot(x_vals, 1/(sigma_true*np.sqrt(2*np.pi))*np.exp(-(x_vals-mu_true)**2/(2*sigma_true**2)),
                  'b-', label='Wahre Normalverteilung')
    axs[0,0].plot(x_vals, 1/(sigma_mle*np.sqrt(2*np.pi))*np.exp(-(x_vals-mu_mle)**2/(2*sigma_mle**2)),
                  'r--', label='MLE Normalverteilung')
    axs[0,0].set_title("Histogram + Wahre & MLE Normalverteilung")
    axs[0,0].legend()
    
    # B: Log-Likelihood μ
    axs[0,1].plot(mu_values, logL_mu, color='green', label='ℓ(μ)')
    axs[0,1].set_title("Log-Likelihood ℓ(μ) bei σ²=MLE")
    axs[0,1].set_xlabel("μ")
    axs[0,1].set_ylabel("ℓ(μ)")
    axs[0,1].axvline(mu_mle, color='red', linestyle='--', label=r'$\hat{\mu}_{MLE}$')
    axs[0,1].legend()
    
    # C: Log-Likelihood σ
    axs[0,2].plot(sigma_values, logL_sigma, color='purple', label='ℓ(σ)')
    axs[0,2].set_title("Log-Likelihood ℓ(σ) bei μ=MLE")
    axs[0,2].set_xlabel("σ")
    axs[0,2].set_ylabel("ℓ(σ)")
    axs[0,2].axvline(sigma_mle, color='red', linestyle='--', label=r'$\hat{\sigma}_{MLE}$')
    axs[0,2].legend()
    
    # D: Datenpunkte vs. μ_MLE
    axs[1,0].plot(data, np.zeros_like(data), 'o', label='Datenpunkte')
    axs[1,0].axvline(mu_mle, color='red', linestyle='--', label=r'$\hat{\mu}_{MLE}$')
    axs[1,0].set_title("Datenpunkte + MLE Mittelwert")
    axs[1,0].legend()
    
    # E: Abweichungen vom Mittelwert
    deviations = data - mu_mle
    axs[1,1].bar(range(n), deviations, color='orange')
    axs[1,1].axhline(0, color='black', linestyle='--')
    axs[1,1].set_title("Abweichungen der Daten vom MLE Mittelwert")
    axs[1,1].set_xlabel("Datenindex")
    axs[1,1].set_ylabel("X_i - μ_MLE")
    
    # F: Sufficient statistics
    axs[1,2].bar([0,1], [np.sum(data), np.sum((data-mu_mle)**2)], color=['blue','green'])
    axs[1,2].set_xticks([0,1])
    axs[1,2].set_xticklabels(['Summe X_i','Summe (X_i-μ_MLE)²'])
    axs[1,2].set_title("Sufficient Statistics für Normalverteilung")
    axs[1,2].set_ylabel("Wert")
    
    plt.tight_layout()
    plt.show()
    
    # Zusammenfassung
    print(f"MLE Mittelwert μ: {mu_mle:.2f}")
    print(f"MLE Standardabweichung σ: {sigma_mle:.2f}")
    print("Sufficient statistics: Summe der Datenpunkte und Summe der quadrierten Abweichungen vom Mittelwert enthalten alle Information für MLE.")
    print("Hinweis: Mit zunehmender Stichprobengröße stabilisiert sich der MLE und nähert sich den wahren Parametern an.")

# Interaktives Widget
interact(
    normal_mle_animated,
    n=IntSlider(min=5, max=20000, step=1, value=1, description="Stichprobengröße"),
    mu_true=FloatSlider(min=-10, max=10, step=0.1, value=5.0, description="Wahrer μ"),
    sigma_true=FloatSlider(min=0.5, max=10, step=0.1, value=2.0, description="Wahre σ")
)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, IntSlider

np.random.seed(42)

def sufficient_stats_widget(mu_true=5.0, sigma_true=2.0, n=30, max_n=200, repetitions=50):
    n_values = np.arange(5, repetitions+1, 5)
    max_n = repetitions
    var_mu_list = []
    var_sigma2_list = []

    # Berechnung Varianz über n
    for n_val in n_values:
        sums_mu = []
        sums_sigma2 = []
        for _ in range(repetitions):
            sample = np.random.normal(mu_true, sigma_true, n_val)
            sums_mu.append(np.sum(sample)/n_val)
            sums_sigma2.append(np.sum((sample - np.mean(sample))**2)/n_val)
        var_mu_list.append(np.var(sums_mu))
        var_sigma2_list.append(np.var(sums_sigma2))

    # Balkendiagramme sufficient statistics für aktuelle n
    sample = np.random.normal(mu_true, sigma_true, n)
    T1_all = np.array([np.sum(np.random.normal(mu_true, sigma_true, n))/n for _ in range(repetitions)])
    T2_all = np.array([np.sum((np.random.normal(mu_true, sigma_true, n) - np.mean(np.random.normal(mu_true, sigma_true, n)))**2)/n for _ in range(repetitions)])

    fig, axs = plt.subplots(3,2, figsize=(16,12))

    # Obere Reihe: Balkendiagramme
    axs[0,0].bar(range(repetitions), T1_all, color='blue')
    axs[0,0].set_title(f"Sufficient Statistic T1 = Mittelwert pro Wiederholung (n={n})")
    axs[0,0].set_xlabel("Wiederholung")
    axs[0,0].set_ylabel("Mittelwert X̄")

    axs[0,1].bar(range(repetitions), T2_all, color='green')
    axs[0,1].set_title(f"Sufficient Statistic T2 = Varianz pro Wiederholung (n={n})")
    axs[0,1].set_xlabel("Wiederholung")
    axs[0,1].set_ylabel("Σ(X_i - X̄)² / n")

    # Mittlere Reihe: Varianz μ̂
    axs[1,0].plot(n_values, var_mu_list, '-o', color='blue')
    axs[1,0].set_title("Varianz des Mittelwerts μ̂ über n")
    axs[1,0].set_xlabel("Stichprobengröße n")
    axs[1,0].set_ylabel("Varianz(μ̂)")

    axs[1,1].scatter(n_values, var_mu_list, color='blue')
    axs[1,1].set_title("Varianz des Mittelwerts μ̂ (Scatter)")
    axs[1,1].set_xlabel("n")
    axs[1,1].set_ylabel("Varianz(μ̂)")

    # Untere Reihe: Varianz σ̂²
    axs[2,0].plot(n_values, var_sigma2_list, '-o', color='green')
    axs[2,0].set_title("Varianz der Varianz σ̂² über n")
    axs[2,0].set_xlabel("Stichprobengröße n")
    axs[2,0].set_ylabel("Varianz(σ̂²)")

    axs[2,1].scatter(n_values, var_sigma2_list, color='green')
    axs[2,1].set_title("Varianz der Varianz σ̂² (Scatter)")
    axs[2,1].set_xlabel("n")
    axs[2,1].set_ylabel("Varianz(σ̂²)")

    plt.tight_layout()
    plt.show()

    print("Interpretation:")
    print("- Obere Balken: Sufficient statistics T1 und T2 über Wiederholungen für gewählte n.")
    print("- Mittlere/untere Zeile: Varianz der MLE-Parameter über verschiedene n-Werte.")
    print("- Sinkende Varianz → stabilere Parameter → robustes Modell.")

# Interaktives Widget
interact(
    sufficient_stats_widget,
    mu_true=IntSlider(min=-10, max=10, step=1, value=5, description="Wahrer μ"),
    sigma_true=IntSlider(min=1, max=10, step=1, value=2, description="Wahre σ"),
    n=IntSlider(min=5, max=200, step=5, value=30, description="Aktuelle Stichprobengröße"),
    max_n=IntSlider(min=1, max=500, step=1, value=1, description="Max n für Varianzplot"),
    repetitions=IntSlider(min=10, max=200, step=10, value=50, description="Wiederholungen")
)
```

## Stabilität der Sufficient Statistics

- **T1 = Summe der Datenpunkte:** enthält alle Information über den Mittelwert μ.  
- **T2 = Summe der quadrierten Abweichungen vom Mittelwert:** enthält alle Information über die Varianz σ².  

**Interpretation der Balken:**
- Schwanken die Balken stark → MLE schwankt, Modell ist weniger robust.  
- Balken sind stabil → sufficient statistics sind konsistent, MLE stabil → robustes Modell.  

> Beobachtung: Mit steigender Stichprobengröße werden die Balken stabiler, die Schätzung von μ und σ² verlässlicher, und das Modell repräsentiert die Population gut.
