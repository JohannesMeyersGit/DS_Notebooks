# Frequentistisch vs. Bayesianisch

## 1. Frequentistische Statistik

**Kernidee:**  
- Parameter einer Population (z. B. Mittelwert $\mu$, Varianz $\sigma^2$) sind **fix, unbekannt und unveränderlich**.  
- Wahrscheinlichkeiten beziehen sich **auf Daten**, nicht auf Parameter.

**Beispiel:**  
- Wir haben eine Münze mit unbekannter Wahrscheinlichkeit für „Kopf“ $p$.  
- Wir werfen die Münze $n=10$ Mal und beobachten $k=7$ Köpfe.  

**Frequentistische Sicht:**  
- $p$ ist **ein fester Wert**, den wir nicht kennen.  
- Die Wahrscheinlichkeit $P(k=7 \mid p)$ beschreibt **die Wahrscheinlichkeit, die Daten zu erhalten**, wenn $p$ tatsächlich dieser feste Wert ist.  
- Maximum-Likelihood-Schätzung (MLE):  

$$
\hat{p}_{MLE} = \frac{k}{n} = \frac{7}{10} = 0.7
$$

- Konfidenzintervalle beschreiben, **wie stabil die Schätzung über viele wiederholte Stichproben wäre**, nicht die Wahrscheinlichkeit, dass $p$ selbst in einem Intervall liegt.

**Hinweis:** Parameter sind fix, Daten variabel. Wahrscheinlichkeiten beziehen sich auf das Zufallsexperiment (Daten), nicht auf $p$.

---

## 2. Bayesianische Statistik

**Kernidee:**  
- Parameter sind **zufällige Variablen**, die wir mit Wahrscheinlichkeiten beschreiben.  
- Wir kombinieren unser **Vorwissen (Prior)** mit den **Daten (Likelihood)**, um die **Posterior-Verteilung** zu erhalten.  

**Bayes-Formel:**  

$$
P(\theta \mid \text{Daten}) = \frac{P(\text{Daten} \mid \theta) \cdot P(\theta)}{P(\text{Daten})}
$$

- $P(\theta)$ = Prior → Wissen oder Annahmen über $\theta$ **vor der Beobachtung der Daten**  
- $P(\text{Daten} \mid \theta)$ = Likelihood → Wahrscheinlichkeit der Daten für einen gegebenen Parameterwert  
- $P(\theta \mid \text{Daten})$ = Posterior → **aktualisiertes Wissen über $\theta$ nach Beobachtung der Daten**  
- $P(\text{Daten})$ = Normierungsfaktor

**Beispiel:** Münze mit Priorwissen  
- Prior: $p \sim \text{Beta}(2,2)$ (leichte Konzentration um $0.5$)  
- Beobachtung: $n=10$ Würfe, $k=7$ Köpfe  
- Posterior:

$$
p \mid k \sim \text{Beta}(\alpha + k, \beta + n - k) = \text{Beta}(2+7, 2+3) = \text{Beta}(9,5)
$$

- Daraus können wir z. B. **MAP-Schätzer**, **erwartetes $p$** oder **credible intervals** ableiten.

**Hinweis:** Parameter sind jetzt zufällig und wir können Wahrscheinlichkeitsaussagen über sie treffen.

---

## 3. Vergleich Frequentistisch vs. Bayesianisch

| Aspekt | Frequentistisch | Bayesianisch |
|--------|----------------|--------------|
| Parameter | Fix, unbekannt | Zufällig, mit Wahrscheinlichkeitsverteilung |
| Wahrscheinlichkeit | Daten | Parameter |
| Vorwissen | Nicht genutzt | Explizit über Prior eingebunden |
| Schätzung | MLE, Konfidenzintervall | Posterior, MAP, Erwartungswert |
| Aussage über Parameter | „Hätte ich unendlich viele Daten, würde das Intervall 95 % der Stichproben enthalten“ | „Die Wahrscheinlichkeit, dass der Parameter in diesem Intervall liegt, beträgt 95 %“ |
| Beispiel | $\hat{p} = 7/10$ | $p \sim \text{Beta}(9,5)$ |

---

## 4. Intuitive Vorstellung

1. **Frequentist:** Parameter ist ein **geheimer, fester Wert**, den wir raten.  
2. **Bayesian:** Parameter ist eine **verteilte Größe**, die wir **mit Wahrscheinlichkeiten beschreiben** und kontinuierlich aktualisieren.  

**Bildliche Analogie:**  
- Frequentist: Du hast einen verschlossenen Beutel mit Murmeln (Parameter), du ziehst Kugeln (Daten) und schätzt den Anteil rot.  
- Bayesian: Du hast eine Vermutung, wie viele rote Murmeln drin sind (Prior), ziehst Kugeln (Daten), und aktualisierst deine Vermutung → Posterior.

---

## 5. Zusammenhang zu MLE / Sufficient Statistics

- **Frequentistisch:** MLE nutzt sufficient statistics, um Parameter punktgenau zu schätzen.  
- **Bayesianisch:** Posterior kann auch sufficient statistics nutzen, aber zusätzlich werden **Wahrscheinlichkeitsverteilungen für die Parameter** generiert.

**Hinweis:** Frequentistische Schätzungen liefern **Punktwerte**, Bayesianische Schätzungen liefern **Verteilungen**, die unser Wissen über den Parameter repräsentieren.


# Bayesianische Statistik Basics

Bayesianische Statistik erlaubt es, **Vorwissen (Prior)**, **Daten (Likelihood)** und **Posterior-Wissen** über Parameter zu kombinieren. Sie ist zentral für probabilistische Modellierung, Unsicherheitsabschätzung und Bayes’sches Lernen in Machine Learning.

---


## 1. Prior-Verteilungen

**Definition:**  
$P(\theta)$ beschreibt unser Wissen oder unsere Annahmen über einen Parameter $\theta$, **vor Beobachtung der Daten**.  

---

### Bedeutung des Priors

- Der Prior **beeinflusst den Posterior**, besonders bei kleinen Stichproben.  
- Je mehr Daten vorhanden sind, desto **dominanter wird die Likelihood** und desto kleiner der Einfluss des Priors.  
- Intuition: Prior = „was wir vor den Daten glaubten“, Likelihood = „was die Daten sagen“, Posterior = „aktualisiertes Wissen“.

**Beispiel Bernoulli-Münze:**  

- Vorwissen: Münze ist wahrscheinlich fair → Prior $p \sim \text{Beta}(2,2)$  
- Beobachtung: $n=3$ Würfe, $k=3$ Köpfe → Posterior: $p \sim \text{Beta}(2+3, 2+0) = \text{Beta}(5,2)$  
- Hier zeigt sich: Bei wenigen Daten beeinflusst der Prior die Posterior-Verteilung deutlich.

---

### Typen von Priors

#### 1. Conjugate Priors

- **Definition:** Ein Prior ist **konjugiert**, wenn der Posterior dieselbe Verteilungsfamilie wie der Prior hat.  
- Vorteil: **Analytisch lösbar**, einfache Berechnung ohne Sampling.  
- Beispiel:  
  - Bernoulli-Daten $X \sim \text{Bernoulli}(p)$  
  - Beta-Prior: $p \sim \text{Beta}(\alpha, \beta)$  
  - Posterior: $p \mid X \sim \text{Beta}(\alpha + k, \beta + n - k)$  
- Intuition: Der Posterior bleibt „im gleichen Format“, nur die Parameter werden angepasst.  

#### 2. Noninformative / Flat Priors

- **Definition:** Priors, die **so wenig Wissen wie möglich einbringen**.  
- Beispiel: Uniformverteilung $p \sim \text{Uniform}(0,1)$ für eine Münze → alle Werte sind a priori gleich wahrscheinlich.  
- Effekt: Daten dominieren den Posterior, der Prior trägt kaum bei.  
- Vorteil: Nützlich, wenn wenig Vorwissen vorhanden ist, oder um „objektive“ Bayesianische Analysen durchzuführen.  
- Hinweis: Noninformative Priors sind besonders bei kleinen Stichproben riskant, da Posterior stark von der Likelihood abhängt.

---

### Effekt des Priors im Bayesian ML

- **Kleine Stichproben:** Prior hat großen Einfluss → kann Wissen aus Expertenwissen oder historischen Daten einfließen lassen.  
- **Große Stichproben:** Prior wird weniger relevant, Posterior wird fast ausschließlich durch die Daten bestimmt.  
- **Conjugate Priors:** Rechenaufwand minimieren, schnelle analytische Updates.  
- **Noninformative Priors:** Objektive Schätzung, Daten dominieren.  

> Merksatz: „Prior + Likelihood = Posterior“ – je stärker die Daten (n groß), desto kleiner der Einfluss des Priors.  

---

## 2. Posterior

**Definition:**  
$P(\theta \mid \text{Daten})$ ist die **aktualisierte Verteilung** über $\theta$ nach Beobachtung der Daten.

**Formel (Bayes-Theorem):**

$$
P(\theta \mid \text{Daten}) = \frac{P(\text{Daten} \mid \theta) \cdot P(\theta)}{P(\text{Daten})}
$$

- $P(\theta)$ = Prior  
- $P(\text{Daten} \mid \theta)$ = Likelihood  
- $P(\text{Daten})$ = Normierung, damit Posterior eine Wahrscheinlichkeitsverteilung ist

**Beispiel (Bernoulli):**  
- Prior: $p \sim \text{Beta}(2,2)$  
- Beobachtung: $n=10$ Würfe, $k=7$ Köpfe  
- Posterior:

$$
p \mid k \sim \text{Beta}(\alpha + k, \beta + n - k) = \text{Beta}(2+7, 2+3) = \text{Beta}(9,5)
$$

**Bezug zu Data Science / ML:**  
- Posterior liefert **eine ganze Verteilung über Parameter**, nicht nur einen Punktwert.  
- Hilfreich z. B. für probabilistische Klassifikation, Bayesian Neural Networks oder Bayes’sche Regressionsmodelle.

---

## 3. MAP – Maximum a Posteriori

**Definition:**  
- Der Wert von $\theta$, der die Posterior-Wahrscheinlichkeit maximiert:

$$
\hat{\theta}_{MAP} = \arg\max_\theta P(\theta \mid \text{Daten})
$$

- **Analogie zur MLE:** Wenn der Prior uniform ist, entspricht MAP = MLE.

**Beispiel (Beta-Bernoulli):**  
- Posterior $p \sim \text{Beta}(9,5)$  
- MAP-Schätzer:

$$
\hat{p}_{MAP} = \frac{\alpha-1}{\alpha + \beta -2} = \frac{9-1}{9+5-2} = \frac{8}{12} = 0.667
$$

**Bezug zu Data Science / ML:**  
- MAP-Schätzer liefert **robuste Punktwerte**, besonders bei kleinen Datenmengen oder informativen Priors.

---

## 4. Credible Intervals

**Definition:**  
- Bayesianisches Gegenstück zu Konfidenzintervallen.  
- Beschreibt die **direkte Wahrscheinlichkeit**, dass ein Parameter in einem Intervall liegt.

**Beispiel:**  
- 95 %-Credible Interval für Posterior $p \sim \text{Beta}(9,5)$ → Intervall [0.47, 0.86]  
- Interpretation: „Mit 95 % Wahrscheinlichkeit liegt $p$ in diesem Intervall.“

**Bezug zu Data Science / ML:**  
- Bietet direkte Aussagen über Unsicherheit von Parametern.  
- Hilfreich bei probabilistischen Vorhersagen, z. B. für Risikoabschätzung oder Bayesian Optimization.

---

## 5. Bayesian Curve Fitting / Regression

**Idee:**  
- Bayesianisches Update auf Regressionsparameter $\theta$:
  - Vorwissen → Prior $P(\theta)$  
  - Daten → Likelihood $P(\text{Daten} \mid \theta)$  
  - Nach Update → Posterior $P(\theta \mid \text{Daten})$  

**Beispiel (lineare Regression):**  
- Daten: $(x_i, y_i)$  
- Modell: $y_i = \theta_0 + \theta_1 x_i + \epsilon$, $\epsilon \sim \mathcal{N}(0, \sigma^2)$  
- Prior: z. B. $\theta_0, \theta_1 \sim \mathcal{N}(0,10)$  
- Posterior: Verteilung der Koeffizienten nach Beobachtung der Daten  
- Output: Posterior-Mittelwerte, MAP, Credible Intervals für $\theta_0, \theta_1$

**Bezug zu Data Science / ML:**  
- Bayesian Regression modelliert **Unsicherheit in den Parametern**, nützlich bei kleinen Datenmengen oder stark verrauschten Daten.  
- Posterior kann für **predictive distributions** genutzt werden: Vorhersagen werden probabilistisch statt punktgenau.

---

### Zusammenfassung

1. **Prior:** Vorwissen über Parameter.  
2. **Likelihood:** Wahrscheinlichkeit der Daten bei gegebenem Parameter.  
3. **Posterior:** Aktualisiertes Wissen über Parameter nach Daten.  
4. **MAP:** Punktwert des Posterior, ähnlich MLE.  
5. **Credible Intervals:** Bayesianisches Unsicherheitsintervall.  
6. **Bayesian Regression / Curve Fitting:** Unsicherheit in Parametern wird direkt modelliert.

**Hinweis:** Bayesianische Statistik erlaubt es, **Punkt- und Verteilungsinformationen zu kombinieren**, priorbasiertes Lernen zu betreiben und Unsicherheit explizit zu quantifizieren – zentral für probabilistisches Machine Learning.


# Bayesianisches Update für Normalverteilung – Step by Step

Wir betrachten eine Normalverteilung und berechnen den Posterior für den Mittelwert $\mu$.

---

## Schritt 0: Problemstellung

- Messdaten: $x_1, \dots, x_n \sim \mathcal{N}(\mu, \sigma^2)$, $\sigma^2$ bekannt  
- Ziel: Posterior für $\mu$  
- Prior: $\mu \sim \mathcal{N}(\mu_0, \sigma_0^2)$  

---

## Schritt 1: Likelihood

Die Likelihood der Daten gegeben $\mu$ ist:

$$
P(x_1, \dots, x_n \mid \mu) = \prod_{i=1}^n \frac{1}{\sqrt{2\pi \sigma^2}} \exp\Big(-\frac{(x_i - \mu)^2}{2\sigma^2}\Big)
$$

- Produkt der Wahrscheinlichkeiten jeder Beobachtung.

---

## Schritt 2: Posterior (konjugierter Fall)

Mit Normalprior $\mu \sim \mathcal{N}(\mu_0, \sigma_0^2)$ ergibt sich ein **Normalposterior**:

$$
\mu \mid x_1,\dots,x_n \sim \mathcal{N}(\mu_n, \sigma_n^2)
$$

mit

$$
\sigma_n^2 = \frac{1}{\frac{n}{\sigma^2} + \frac{1}{\sigma_0^2}}, \quad
\mu_n = \sigma_n^2 \left( \frac{\sum_{i=1}^n x_i}{\sigma^2} + \frac{\mu_0}{\sigma_0^2} \right)
$$

**Interpretation:**

- $\sigma_n^2$ sinkt mit zunehmender Stichprobengröße → mehr Daten, mehr Sicherheit  
- $\mu_n$ = gewichtetes Mittel aus Prior-Mittelwert $\mu_0$ und Stichprobenmittelwert $\bar{x}$

---

## Schritt 3: Numerisches Beispiel

- Prior: $\mu_0 = 5$, $\sigma_0^2 = 4$  
- Daten: $x = [4.8, 5.2, 5.5]$, $\sigma^2 = 1$  

1. Berechnung: $n/\sigma^2 + 1/\sigma_0^2 = 3 + 0.25 = 3.25$  
2. Posterior-Varianz: $\sigma_n^2 = 1/3.25 \approx 0.3077$  
3. Posterior-Mittelwert:

$$
\mu_n = 0.3077 \cdot \left(\frac{4.8+5.2+5.5}{1} + \frac{5}{4}\right) = 0.3077 \cdot 16.75 \approx 5.15
$$

**Ergebnis:**

$$
\mu \mid x \sim \mathcal{N}(5.15, 0.3077)
$$

---

## Schritt 4: MAP & 95%-Credible Interval

- MAP = Posterior-Mittelwert: $\hat{\mu}_{MAP} = 5.15$  
- 95 %-Credible Interval:

$$
\mu_n \pm 1.96 \cdot \sigma_n \approx 5.15 \pm 1.96 \cdot 0.555 \approx [4.05, 6.25]
$$

- Interpretation: „Mit 95 % Wahrscheinlichkeit liegt $\mu$ in diesem Intervall.“

---

## Schritt 5: Interpretation

- Posterior-Mittelwert liegt zwischen Prior-Mittelwert (5) und Stichprobenmittelwert (~5.17), gewichtet nach Unsicherheit  
- Posterior-Varianz < Prior-Varianz → Daten haben Unsicherheit reduziert  
- Credible Interval gibt direkte Wahrscheinlichkeit für $\mu$  

**Bezug zu Data Science / ML:**  
- Posterior liefert **ganze Verteilung** über Parameter → probabilistische Vorhersagen möglich  
- MAP-Schätzer liefert **robuste Punktwerte**  
- Bayesianisches Update kann in probabilistischer Regression oder Bayesian Neural Networks verwendet werden


## Effekt der Prior auf die Modellguete

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider, IntSlider, Button, VBox, Output

np.random.seed(42)
out = Output()
state = {'x': None}

def bayesian_normal_prior_comparison_map(n=5, mu_true=5.0, sigma_true=2.0, mu0=4.0, sigma0=1.0, repeat=False):
    if repeat or state['x'] is None:
        state['x'] = np.random.normal(mu_true, sigma_true, n)
    x = state['x']
    x_bar = np.mean(x)
    
    theta = np.linspace(mu_true-4*sigma_true, mu_true+4*sigma_true, 200)
    likelihood = np.exp(-0.5 * n * ((theta - x_bar)/sigma_true)**2)
    likelihood /= likelihood.sum()
    
    priors = {
        'informativ': np.exp(-0.5*((theta - mu0)/sigma0)**2),
        'semi-informativ': np.exp(-0.5*((theta - mu0)/(sigma0*2))**2),
        'noninformativ': np.ones_like(theta)
    }
    
    fig, axs = plt.subplots(1,3, figsize=(18,5))
    
    for i, (key, prior) in enumerate(priors.items()):
        prior /= prior.sum()
        posterior = prior * likelihood
        posterior /= posterior.sum()
        theta_map = theta[np.argmax(posterior)]
        cumsum = np.cumsum(posterior)
        lower = theta[np.searchsorted(cumsum, 0.025)]
        upper = theta[np.searchsorted(cumsum, 0.975)]
        
        axs[i].plot(theta, prior, label='Prior', color='orange')
        axs[i].plot(theta, likelihood, label='Likelihood', color='green')
        axs[i].plot(theta, posterior, label='Posterior', color='blue')
        axs[i].axvline(theta_map, color='red', linestyle='--', label=f'MAP={theta_map:.2f}')
        axs[i].axvline(mu_true, color='purple', linestyle=':', label=f'Wahrer μ={mu_true}')
        axs[i].axvspan(lower, upper, color='blue', alpha=0.2, label='95% CI')
        axs[i].scatter(x, np.zeros_like(x), color='black', marker='x', s=50, label='Daten')
        axs[i].set_title(f"{key} Prior\nAbweichung MAP-Wert: {theta_map-mu_true:+.2f}")
        axs[i].set_xlabel(r'$\mu$')
        axs[i].set_ylabel('Dichte (normiert)')
        axs[i].legend()
    
    plt.tight_layout()
    plt.show()
    
    with out:
        out.clear_output(wait=True)
        print(f"Stichprobenmittelwert: {x_bar:.2f}, n={n}")
        print("Beobachte, wie der Prior die Posterior-Verteilung beeinflusst:")
        print("- Informativ: starker Einfluss, Posterior nahe Prior-Mittelwert")
        print("- Semi-informativ: mittlerer Einfluss")
        print("- Noninformativ: Posterior fast nur durch Daten bestimmt")
        print("- MAP vs. wahrer Wert zeigt Abweichung (Bias) durch Prior und Stichprobe")

repeat_button = Button(description="Neue Stichprobe ziehen", button_style='info')
def on_repeat_button_clicked(b):
    bayesian_normal_prior_comparison_map(repeat=True)
repeat_button.on_click(on_repeat_button_clicked)

interact(
    bayesian_normal_prior_comparison_map,
    n=IntSlider(min=1, max=50, step=1, value=5, description="Stichprobengröße"),
    mu_true=FloatSlider(min=-10, max=10, step=0.1, value=5, description="Wahrer μ"),
    sigma_true=FloatSlider(min=0.1, max=5, step=0.1, value=2, description="Wahre σ"),
    mu0=FloatSlider(min=-10, max=10, step=0.1, value=4, description="Prior μ₀"),
    sigma0=FloatSlider(min=0.1, max=5, step=0.1, value=1, description="Prior σ₀")
)

VBox([repeat_button, out])
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider, IntSlider, Dropdown

np.random.seed(42)

def sequential_bayesian_update_sigma(mu_true=5.0, sigma_true=2.0, n=20, prior_type='informativ'):
    # Prior definieren
    if prior_type == 'informativ':
        mu_prior = mu_true*0.8
        sigma_prior = sigma_true*0.5
        sigma0 = sigma_true*0.5
    elif prior_type == 'semi-informativ':
        mu_prior = 0
        sigma_prior = 1*2
        sigma0 = 1*2
    else:
        mu_prior = 0
        sigma_prior = 1e6
        sigma0 = 1e6
    
    # Stichprobe
    x = np.random.normal(mu_true, sigma_true, n)
    
    # Arrays für Tracking
    mu_post_list = []
    sigma_post_list = []
    
    # Theta für Posterior-Plot
    theta = np.linspace(mu_true-4*sigma_true, mu_true+4*sigma_true, 200)
    posterior_plot_list = []
    
    mu_post, sigma_post = mu_prior, sigma_prior
    for i, xi in enumerate(x, 1):
        # Posterior für μ analytisch
        sigma2_post = 1 / (1/sigma_post**2 + 1/sigma_true**2)
        mu_post = sigma2_post * (mu_post/sigma_post**2 + xi/sigma_true**2)
        sigma_post = np.sqrt(sigma2_post)
        
        mu_post_list.append(mu_post)
        sigma_post_list.append(sigma_post)
        
        # Posterior-Dichte für Plot
        posterior = (1/(sigma_post*np.sqrt(2*np.pi))) * np.exp(-0.5*((theta-mu_post)/sigma_post)**2)
        posterior_plot_list.append(posterior)
    
    # Plot
    fig, axs = plt.subplots(3,1, figsize=(10,14))
    
    # Oben: Posterior-Dichte nach jedem Sample
    axs[0].plot(theta, (1/(sigma_prior*np.sqrt(2*np.pi))) * np.exp(-0.5*((theta-mu_prior)/sigma_prior)**2), color='orange', label='Initial Prior')
    for i in range(n):
        axs[0].plot(theta, posterior_plot_list[i], color=plt.cm.viridis(i/n), alpha=0.6)
    axs[0].axvline(mu_true, color='purple', linestyle=':', label='Wahrer μ')
    axs[0].set_title("Sequentielle Posterior-Updates μ")
    axs[0].set_xlabel(r'$\mu$')
    axs[0].set_ylabel("Dichte")
    axs[0].legend()
    
    # Mittig: MAP μ über n
    axs[1].plot(range(1,n+1), mu_post_list, '-o', color='red', label='MAP μ')
    axs[1].axhline(mu_true, color='purple', linestyle=':', label='Wahrer μ')
    axs[1].set_xlabel("Anzahl Samples")
    axs[1].set_ylabel("MAP μ")
    axs[1].set_title("MAP μ über sequentielle Updates")
    axs[1].legend()
    
    # Unten: MAP σ über n
    axs[2].plot(range(1,n+1), sigma_post_list, '-o', color='green', label='MAP σ')
    axs[2].axhline(sigma_true, color='purple', linestyle=':', label='Wahre σ')
    axs[2].set_xlabel("Anzahl Samples")
    axs[2].set_ylabel("MAP σ")
    axs[2].set_title("MAP σ über sequentielle Updates")
    axs[2].legend()
    
    plt.tight_layout()
    plt.show()
    
    print("Interpretation:")
    print("- Posterior für μ wird Schritt für Schritt aktualisiert.")
    print("- MAP μ und MAP σ nähern sich mit zunehmenden Samples den wahren Werten.")
    print("- Die Posterior-Varianz nimmt ab → Sättigung wird sichtbar, zusätzliche Daten liefern kaum neue Information.")

interact(
    sequential_bayesian_update_sigma,
    mu_true=FloatSlider(min=-10, max=10, step=0.1, value=5, description="Wahrer μ"),
    sigma_true=FloatSlider(min=0.1, max=5, step=0.1, value=2, description="Wahre σ"),
    n=IntSlider(min=1, max=1000, step=1, value=1, description="Samples"),
    prior_type=Dropdown(options=['informativ', 'semi-informativ', 'noninformativ'], value='informativ', description='Prior-Typ')
)
```

## Bayesianisches Posterior-Widget für Normalverteilung

Dieses interaktive Widget visualisiert den Einfluss der Stichprobengröße und des Priors auf den Posterior einer Normalverteilung.  

### Funktionen des Widgets

1. **Samples ziehen**  
   - Ein Slider für die Stichprobengröße \(n\). Mit wachsendem \(n\) werden mehr Datenpunkte gezogen, und du kannst live beobachten, wie sich Prior, Likelihood und Posterior verändern.

2. **Prior-Typ wählen**  
   - Dropdown-Menü mit:
     - **informativ**: starker Einfluss des Priors, Posterior wird stärker in Richtung Prior gezogen, besonders bei kleinen \(n\).  
     - **semi-informativ**: mittlerer Einfluss des Priors.  
     - **noninformativ**: kaum Einfluss des Priors, Posterior wird fast ausschließlich von den Daten bestimmt.

3. **Posterior visualisieren**  
   - Blau: Posterior  
   - Orange: Prior  
   - Grün: Likelihood der aktuellen Stichprobe  
   - Rote Linie: MAP (Maximum A Posteriori)  
   - Violette Linie: wahrer Parameter \(\mu\)  
   - Hellblauer Bereich: 95%-Credible Interval

4. **Posterior-Varianz über n**  
   - Unterer Subplot zeigt die Varianz des Posterior für jede gewählte Stichprobengröße.  
   - Sinkende Varianz → Posterior stabilisiert sich → MAP wird robust  
   - Ab einem bestimmten \(n\) flacht die Kurve ab: **Sättigung von Informationen** erreicht. Weitere Daten liefern kaum neue Information.

### Interpretation

- **MAP-Konvergenz:** Mit wachsendem \(n\) nähert sich die MAP dem wahren Wert \(\mu\) an.  
- **Credible Interval:** Die Unsicherheit nimmt ab, die Posterior-Dichte wird schmaler.  
- **Sättigung der Informationen:** Sobald die Posterior-Varianz praktisch konstant wird, ist der Parameter robust geschätzt; zusätzliche Daten verändern Posterior und MAP nur minimal.

### Bezug zu Data Science / ML

- Bayesianische Schätzungen sind besonders nützlich bei **kleinen Datensätzen** oder wenn **Vorwissen** (Prior) integriert werden soll.  
- Die Sättigung zeigt, ab welcher Stichprobengröße der Schätzer stabil ist, was **Ressourcen spart** und die Notwendigkeit weiterer Daten verdeutlicht.  

> Tipp: Experimentiere mit unterschiedlichen Prior-Typen und Stichprobengrößen, um den Effekt von Priorwissen und Datenmenge auf den Posterior zu verstehen.
