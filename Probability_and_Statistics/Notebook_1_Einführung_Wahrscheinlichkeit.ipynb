## 1. Grundbegriffe

### Zufallsexperimente
Ein Zufallsexperiment ist ein Vorgang, dessen Ergebnis nicht sicher vorhergesagt werden kann, aber alle möglichen Ergebnisse bekannt sind.  
**Beispiele:**
- Werfen eines Würfels
- Ziehen einer Karte aus einem Kartenspiel
- Ziehen einer Kugel aus einer Urne

### Ergebnisraum / Stichprobenraum
Der **Ergebnisraum** (oder Stichprobenraum) ist die Menge aller möglichen Ergebnisse eines Zufallsexperiments.  
- **Würfel:** {1, 2, 3, 4, 5, 6}  
- **Karte aus einem Standarddeck:** {Herz A, Herz 2, …, Pik K}  
- **Urne mit roten und blauen Kugeln:** {rot, blau}

### Ereignisse
Ein **Ereignis** ist eine Teilmenge des Ergebnisraums. Es kann aus einem oder mehreren Ergebnissen bestehen.  
- **Einfaches Ereignis:** Eine 6 beim Würfeln → {6}  
- **Zusammengesetztes Ereignis:** Gerade Zahl beim Würfeln → {2, 4, 6}

### Elementarereignisse
Elementarereignisse sind die kleinsten möglichen Ergebnisse, die nicht weiter unterteilt werden können.  
- Würfelwurf → Elementarereignisse: 1, 2, 3, 4, 5, 6  

---

### Einfaches 1D-Beispiel (Würfel)
| Wurf | 1 | 2 | 3 | 4 | 5 | 6 |
|------|---|---|---|---|---|---|
| Ergebnisraum | ✓ | ✓ | ✓ | ✓ | ✓ | ✓ |
| Ereignis A (gerade) |   | ✓ |   | ✓ |   | ✓ |

**Beschreibung:**  
- Ergebnisraum = alle Zahlen von 1 bis 6  
- Ereignis „gerade Zahl“ = Teilmenge {2, 4, 6}  
- Elementarereignisse = jede einzelne Zahl  


````python
import matplotlib.pyplot as plt
from ipywidgets import interact, Dropdown

# Ergebnisraum Würfel
ergebnisse = [1, 2, 3, 4, 5, 6]

# Funktion zur interaktiven Visualisierung
def visualize_event(ereignis_typ):
    # Ereignis auswählen
    if ereignis_typ == "gerade Zahlen":
        ereignis = [2, 4, 6]
    elif ereignis_typ == "ungerade Zahlen":
        ereignis = [1, 3, 5]
    elif ereignis_typ == "eine Zahl (6)":
        ereignis = [6]
    else:
        ereignis = []
    
    plt.figure(figsize=(8,1))
    
    # Punkte zeichnen
    for e in ergebnisse:
        if e in ereignis:
            plt.plot(e, 0, 'o', markersize=20, color='green')
        else:
            plt.plot(e, 0, 'o', markersize=20, color='lightgray')
    
    plt.yticks([])
    plt.xticks(ergebnisse)
    plt.xlim(0.5, 6.5)
    plt.title("Ergebnisraum (Ω) mit Ereignis A")
    
    # Dynamische Legende
    plt.scatter([], [], color='green', label='Ereignis')
    plt.scatter([], [], color='lightgray', label='Ω')
    plt.legend(loc='center left', bbox_to_anchor=(1, 0.5))
    plt.tight_layout()
    plt.show()

# Dropdown-Widget
interact(visualize_event, ereignis_typ=Dropdown(
    options=["gerade Zahlen", "ungerade Zahlen", "eine Zahl (6)"],
    value="gerade Zahlen",
    description="Ereignis auswählen:"
));
````

## 2. Wahrscheinlichkeitsaxiome

Die Wahrscheinlichkeitsrechnung basiert auf den **Kolmogorov-Axiomen**. Sie bilden die mathematische Grundlage, um konsistent mit Wahrscheinlichkeiten zu rechnen.

### 1. Nicht-Negativität
Jedes Ereignis hat eine Wahrscheinlichkeit größer oder gleich null:
\[
P(E) $\ge$ 0
\]

**Beispiel:**  
- Würfeln: P(Zahl = 3) = 1/6 ≥ 0

### 2. Normierung
Die Wahrscheinlichkeit des gesamten Ergebnisraums Ω ist 1:
\[
P(\Omega) = 1
\]

**Beispiel:**  
- Würfelwurf: Ω = {1,2,3,4,5,6}  
- Summe aller Wahrscheinlichkeiten: 6 × 1/6 = 1

### 3. Additivität für disjunkte Ereignisse
Für zwei **disjunkte** Ereignisse A und B (d. h. A ∩ B = ∅) gilt:
\[
P(A $\cup$ B) = P(A) + P(B)
\]

**Beispiel:**  
- A = gerade Zahl {2,4,6}  
- B = Primzahl {2,3,5} → nicht disjunkt, also allgemeiner:  
\[
P(A $\cup$ B) = P(A) + P(B) - P(A $\cap$ B)
\]

---

### Konsequenzen der Axiome
- Wahrscheinlichkeit liegt immer zwischen 0 und 1: 0 ≤ P(E) ≤ 1  
- Leeres Ereignis: P(∅) = 0  
- Wahrscheinlichkeiten von komplementären Ereignissen: P(A^c) = 1 − P(A)  
- Additionssatz für beliebige Ereignisse: P(A ∪ B) = P(A) + P(B) − P(A ∩ B)

---

````python
import matplotlib.pyplot as plt
from matplotlib_venn import venn2
from ipywidgets import interact, SelectMultiple

# Ergebnisraum
ergebnisse = set(range(1, 7))

# Vordefinierte Ereignisse
ereignis_optionen = {
    "gerade Zahlen": {2, 4, 6},
    "ungerade Zahlen": {1, 3, 5},
    "Primzahlen": {2, 3, 5},
    "eine Zahl (6)": {6}
}

def visualize_venn2(selected_sets):
    if len(selected_sets) != 2:
        print("Bitte genau zwei Ereignisse auswählen.")
        return
    
    A = ereignis_optionen[selected_sets[0]]
    B = ereignis_optionen[selected_sets[1]]
    
    plt.figure(figsize=(6,6))
    v = venn2([A, B], set_labels=selected_sets)
    
    # Elemente in die Teilbereiche schreiben
    subsets = {
        '10': A - B,
        '01': B - A,
        '11': A & B
    }
    
    for idx, elems in subsets.items():
        label = v.get_label_by_id(idx)
        if label is not None:
            text = ','.join(map(str, sorted(elems))) if elems else ''
            label.set_text(text)
    
    # Farben anpassen (Patch prüfen)
    patch_ids = {'10':'lightgreen', '01':'lightblue', '11':'yellow'}
    for pid, color in patch_ids.items():
        patch = v.get_patch_by_id(pid)
        if patch is not None:
            patch.set_color(color)
            patch.set_alpha(0.6 if pid != '11' else 0.8)
    
    plt.title("Interaktives 2-Mengen-Venn mit sichtbaren Elementen")
    plt.show()

interact(
    visualize_venn2,
    selected_sets=SelectMultiple(
        options=list(ereignis_optionen.keys()),
        value=["gerade Zahlen", "Primzahlen"],
        description="Ereignisse:",
        rows=4
    )
)

````

## Interaktives 2-Mengen-Venn-Diagramm

Dieses Diagramm zeigt die grundlegenden Konzepte der Wahrscheinlichkeitsrechnung anhand von zwei Ereignissen:

### Wichtige Beobachtungen

1. **Elemente innerhalb eines Kreises = Ereignis A oder B**
   - Jedes Element entspricht einem möglichen Ergebnis im Ergebnisraum.
   - **Grün** = nur in A, **Blau** = nur in B, **Gelb** = Schnittmenge (A ∩ B).

2. **Disjunkte Ereignisse**
   - Wenn zwei Kreise keine Überlappung haben, sind sie **disjunkt** → P(A ∩ B) = 0.
   - Dies illustriert die **Additivität für disjunkte Ereignisse**:  
     \[
     P(A $\cup$ B) = P(A) + P(B)
     \]

3. **Schnittmengen (Überlappungen)**
   - Gelbe Bereiche zeigen gemeinsame Elemente → P(A ∩ B) > 0.
   - Wichtig für den **allgemeinen Additionssatz**:  
     \[
     P(A $\cup$ B) = P(A) + P(B) - P(A $\cap$ B)
     \]

4. **Normierung**
   - Die Summe aller Elemente im Ergebnisraum entspricht 1, wenn Wahrscheinlichkeiten zugewiesen werden (z. B. für Würfel: jedes Element P=1/6).

5. **Nicht-Negativität**
   - Alle Wahrscheinlichkeiten sind ≥ 0, erkennbar daran, dass jedes Element nur in einer oder mehreren Mengen vorkommt.

## Multiplikationssatz der Wahrscheinlichkeit

Der Multiplikationssatz beschreibt, wie die Wahrscheinlichkeit des **gemeinsamen Eintretens von zwei Ereignissen** berechnet wird.

### 1. Unabhängige Ereignisse
Zwei Ereignisse A und B heißen **unabhängig**, wenn das Eintreten des einen Ereignisses das Eintreten des anderen **nicht beeinflusst**.  
Für unabhängige Ereignisse gilt:

\[
P(A $\cap$ B) = P(A) $\cdot$ P(B)
\]

**Beispiel:**  
- Zwei Würfel werden geworfen.  
- Ereignis A: erster Würfel zeigt eine gerade Zahl {2,4,6}  
- Ereignis B: zweiter Würfel zeigt eine gerade Zahl {2,4,6}  
- P(A ∩ B) = 3/6 * 3/6 = 1/4

---

### 2. Abhängige Ereignisse
Zwei Ereignisse heißen **abhängig**, wenn das Eintreten des einen Ereignisses das Eintreten des anderen beeinflusst.  
Für abhängige Ereignisse gilt:

\[
P(A $\cap$ B) = P(A) $\cdot$ P(B|A)
\]

wobei P(B|A) die **bedingte Wahrscheinlichkeit** von B gegeben A ist.

**Beispiel:**  
- Zwei Karten werden ohne Zurücklegen gezogen.  
- Ereignis A: erste Karte ist Herz  
- Ereignis B: zweite Karte ist Herz  
- P(A ∩ B) = 13/52 * 12/51 ≈ 0,0588

````python
import matplotlib.pyplot as plt
from ipywidgets import interact, ToggleButtons
from itertools import product

# Ereignisse für Würfel (unabhängig)
ergebnisse_wuerfel = list(product(range(1,7), repeat=2))
ereignis_unabh = {
    "erste gerade Zahl": lambda x: x[0] % 2 == 0,
    "zweite gerade Zahl": lambda x: x[1] % 2 == 0
}

# Ereignisse für Karten (abhängig, ohne Zurücklegen)
# 13 Herzkarten, 39 Nicht-Herzkarten
ergebnisse_karten = [(i,j) for i in range(1,53) for j in range(1,53) if j != i]

def is_herz(card):
    return card <= 13  # Karten 1-13 = Herz

def visualize_multiplikation(event_type):
    plt.figure(figsize=(6,6))
    
    if event_type == "Unabhängig (Würfel)":
        ergebnisse = ergebnisse_wuerfel
        A = [e for e in ergebnisse if ereignis_unabh["erste gerade Zahl"](e)]
        B = [e for e in ergebnisse if ereignis_unabh["zweite gerade Zahl"](e)]
        P_A = len(A)/len(ergebnisse)
        P_B = len(B)/len(ergebnisse)
        P_AB = len([e for e in ergebnisse if e in A and e in B])/len(ergebnisse)
        plt.bar(["P(A)","P(B)","P(A∩B)"], [P_A, P_B, P_AB], color=['lightgreen','lightblue','yellow'])
        plt.ylim(0,1)
        plt.title("Unabhängige Ereignisse - Würfel")
        plt.show()
        print(f"P(A) = {P_A:.3f}")
        print(f"P(B) = {P_B:.3f}")
        print(f"P(A ∩ B) = {P_AB:.3f}")
        print(f"P(A)*P(B) = {P_A*P_B:.3f} (Multiplikationssatz)")
        
    else:
        # Abhängige Ereignisse: Karte ziehen ohne Zurücklegen
        ergebnisse = ergebnisse_karten
        A = [e for e in ergebnisse if is_herz(e[0])]
        B = [e for e in A if is_herz(e[1])]  # zweite Karte Herz gegeben erste Herz
        P_A = len(A)/len(ergebnisse)
        P_B_given_A = len(B)/len(A)
        P_AB = P_A * P_B_given_A
        
        plt.bar(["P(A)","P(B|A)","P(A∩B)"], [P_A, P_B_given_A, P_AB], color=['lightgreen','lightblue','yellow'])
        plt.ylim(0,1)
        plt.title("Abhängige Ereignisse - Karten ohne Zurücklegen")
        plt.show()
        
        print(f"P(A) = {P_A:.3f}")
        print(f"P(B|A) = {P_B_given_A:.3f}")
        print(f"P(A ∩ B) = P(A)*P(B|A) = {P_AB:.3f} (Multiplikationssatz)")
        
interact(
    visualize_multiplikation,
    event_type=ToggleButtons(
        options=["Unabhängig (Würfel)","Abhängig (Karten)"],
        description="Ereignistyp:"
    )
)
````

## 5. Beispiele

Weitere Beispiele basierend auf einer Urne aus der blind farbige Kugeln mit und ohne zurücklegen gezogen werden zur Beschreibung bedingter und unbedingter Wahrscheinlichkeiten

````python
import matplotlib.pyplot as plt
from ipywidgets import interact, ToggleButtons, IntSlider, IntText
import random

def visualize_urn(event_type, kugeln_anzahl, ziehen, seed):
    random.seed(seed)  # Reproduzierbarkeit
    
    # Farben: Rot, Blau, Grün, Gelb
    farben_labels = ["Rot", "Blau", "Grün", "Gelb"]
    farben_rgb = ["red", "blue", "green", "yellow"]
    
    # Urne aufbauen
    urne = [farben_labels[i % len(farben_labels)] for i in range(kugeln_anzahl)]
    
    # Theoretische Wahrscheinlichkeiten
    theor_counts = {farbe: urne.count(farbe) for farbe in farben_labels}
    theor_probs = {farbe: theor_counts[farbe]/len(urne) for farbe in farben_labels}
    
    gezogene_kugeln = []
    
    if event_type == "Mit Zurücklegen":
        for _ in range(ziehen):
            gezogene_kugeln.append(random.choice(urne))
    else:  # Ohne Zurücklegen
        urne_copy = urne.copy()
        for _ in range(min(ziehen, len(urne_copy))):
            gezogene = random.choice(urne_copy)
            gezogene_kugeln.append(gezogene)
            urne_copy.remove(gezogene)
    
    # Empirische Häufigkeiten
    counts = {farbe: gezogene_kugeln.count(farbe) for farbe in farben_labels}
    
    # Plot
    plt.figure(figsize=(8,4))
    
    bar_width = 0.35
    index = range(len(farben_labels))
    
    plt.bar([i - bar_width/2 for i in index], [theor_probs[f] for f in farben_labels],
            bar_width, color=farben_rgb, alpha=0.5, label="Theoretisch")
    plt.bar([i + bar_width/2 for i in index], [counts[f]/ziehen if ziehen>0 else 0 for f in farben_labels],
            bar_width, color=farben_rgb, alpha=1.0, label="Empirisch")
    
    plt.xticks(index, farben_labels)
    plt.ylim(0,1)
    plt.ylabel("Wahrscheinlichkeit")
    plt.title(f"Ziehung von {ziehen} Kugeln ({event_type})")
    plt.legend()
    plt.show()
    
    print("Gezogene Kugeln:", gezogene_kugeln)
    print("Empirische Wahrscheinlichkeiten:")
    for farbe in farben_labels:
        print(f"P({farbe}) ≈ {counts[farbe]/ziehen if ziehen>0 else 0:.2f}")
    print("Theoretische Wahrscheinlichkeiten:")
    for farbe in farben_labels:
        print(f"P({farbe}) = {theor_probs[farbe]:.2f}")

interact(
    visualize_urn,
    event_type=ToggleButtons(
        options=["Mit Zurücklegen", "Ohne Zurücklegen"],
        description="Ziehung:"
    ),
    kugeln_anzahl=IntSlider(value=8, min=4, max=20, step=1, description="Kugeln in Urne"),
    ziehen=IntSlider(value=5, min=1, max=20, step=1, description="Anzahl Ziehungen"),
    seed=IntText(value=42, description="Seed")
)
````

## 5. Einführung in Simulationen

Simulationen sind ein mächtiges Werkzeug, um Wahrscheinlichkeiten und Zufallsexperimente **experimentell zu untersuchen**. Dabei wiederholen wir ein Zufallsexperiment viele Male und beobachten die Häufigkeiten der Ergebnisse, um **empirische Wahrscheinlichkeiten** zu schätzen.  

### Konzepte

- **Empirische Wahrscheinlichkeit:**  
  $$
  P_{\rm emp}(E) = \frac{\text{Anzahl der Male, die Ereignis E eintrat}}{\text{Anzahl der Wiederholungen}}
  $$

- **Vergleich mit theoretischer Wahrscheinlichkeit:** Simulationswerte nähern sich bei genügend Wiederholungen den theoretischen Wahrscheinlichkeiten an (Gesetz der großen Zahlen).

### Beispiele mit `numpy.random`

**1. Würfelwurf**
```python
import numpy as np

n_wuerfe = 1000
würfel = np.random.randint(1,7,size=n_wuerfe)
emp_prob_6 = np.mean(würfel == 6)
print(f"Empirische Wahrscheinlichkeit für eine 6: {emp_prob_6:.3f}")


````python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, IntSlider, IntText
import random

def simulate_coin_flips(n_flips, seed):
    random.seed(seed)
    np.random.seed(seed)
    
    outcomes = np.random.choice(["Kopf", "Zahl"], size=n_flips)
    counts = {"Kopf": np.sum(outcomes=="Kopf"), "Zahl": np.sum(outcomes=="Zahl")}
    
    plt.figure(figsize=(5,4))
    plt.bar(counts.keys(), [counts["Kopf"]/n_flips, counts["Zahl"]/n_flips], color=["blue","orange"])
    plt.ylim(0,1)
    plt.ylabel("Empirische Wahrscheinlichkeit")
    plt.title(f"{n_flips} Münzwürfe")
    plt.show()
    
    print("Empirische Wahrscheinlichkeiten:")
    for k,v in counts.items():
        print(f"P({k}) ≈ {v/n_flips:.3f}")

interact(
    simulate_coin_flips,
    n_flips=IntSlider(min=1, max=1000, step=1, value=100, description="Anzahl Würfe"),
    seed=IntText(value=42, description="Seed")
)
````

## Abschlussbeispiele: Geburtstagsproblem & Monty Hall Problem

### 1. Geburtstagsproblem
**Ziel:** Wahrscheinlichkeit, dass in einer Gruppe von n Personen mindestens zwei den gleichen Geburtstag haben.

- **Theorie:**  
  - Unabhängige Zufallsvariablen für Geburtstage  
  - P(kein gemeinsamer Geburtstag) = 365/365 * 364/365 * ...  
  - P(mindestens ein gemeinsamer Geburtstag) = 1 - P(kein gemeinsamer Geburtstag)

- **Python-Simulation:**  
  - Ziehe für jede Person zufällig einen Geburtstag (1–365)  
  - Prüfe, ob es Duplikate gibt  
  - Wiederhole Simulation viele Male, berechne empirische Wahrscheinlichkeit

```python
import numpy as np

def birthday_simulation(n_people=23, n_sim=10000, seed=42):
    np.random.seed(seed)
    count = 0
    for _ in range(n_sim):
        birthdays = np.random.randint(1,366, size=n_people)
        if len(birthdays) != len(set(birthdays)):
            count += 1
    return count / n_sim

p_emp = birthday_simulation()
print(f"Empirische Wahrscheinlichkeit: {p_emp:.3f}")


````python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, IntSlider, IntText

def birthday_widget(n_people=23, n_sim=10000, seed=42):
    np.random.seed(seed)
    count = 0
    for _ in range(n_sim):
        birthdays = np.random.randint(1,366, size=n_people)
        if len(birthdays) != len(set(birthdays)):
            count += 1
    p_emp = count / n_sim
    
    # Ausgabe
    print(f"Empirische Wahrscheinlichkeit für mindestens einen gemeinsamen Geburtstag: {p_emp:.3f}")
    
    # Visualisierung
    plt.figure(figsize=(6,1))
    plt.bar(["P(≥2 gleiche Geburtstage)", "P(keine Übereinstimmung)"], [p_emp, 1-p_emp], color=["orange","lightblue"])
    plt.ylim(0,1)
    plt.ylabel("Wahrscheinlichkeit")
    plt.title(f"{n_people} Personen, {n_sim} Simulationen")
    plt.show()

interact(
    birthday_widget,
    n_people=IntSlider(min=2, max=100, step=1, value=23, description="Personen"),
    n_sim=IntText(value=10000, description="Simulationen"),
    seed=IntText(value=42, description="Seed")
)
````

## Monty Hall Problem

**Beschreibung:**  
Das Monty-Hall-Problem ist ein klassisches Beispiel für bedingte Wahrscheinlichkeiten und zeigt, wie intuitives Denken oft von der tatsächlichen Wahrscheinlichkeit abweicht.

**Szenario:**  
1. Du stehst vor drei verschlossenen Türen. Hinter einer Tür befindet sich ein Preis (z.B. Auto), hinter den anderen Ziegen.  
2. Du wählst eine Tür aus.  
3. Der Moderator, der weiß, was sich hinter den Türen befindet, öffnet eine andere Tür, hinter der sich sicher kein Auto befindet.  
4. Du hast nun die Wahl: bei deiner ursprünglichen Tür bleiben oder zu der noch geschlossenen Tür wechseln.  

**Fragen:**  
- Welche Strategie maximiert die Gewinnwahrscheinlichkeit?  
- Warum erhöht Wechseln die Gewinnchance?  

**Wichtige Konzepte:**  
- Bedingte Wahrscheinlichkeit: Die Wahrscheinlichkeit, dass das Auto hinter der gewählten Tür ist, verändert sich, nachdem der Moderator eine Tür geöffnet hat.  
- Unabhängigkeit vs. Abhängigkeit: Die ursprüngliche Wahl ist unabhängig, aber das Öffnen einer Tür durch den Moderator beeinflusst die bedingte Wahrscheinlichkeit der verbleibenden Türen.

**Übungsaufgabe für Studierende:**  
- Simuliere das Monty-Hall-Problem 10.000 Mal sowohl für „Wechseln“ als auch „Bleiben“.  
- Berechne die Gewinnwahrscheinlichkeit für beide Strategien.  
- Diskutiere, warum die intuitive Vermutung oft falsch ist.


````python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, IntSlider, Checkbox, IntText

def monty_hall_sim(n_sim=10000, switch=True, seed=42):
    np.random.seed(seed)
    wins = 0
    for _ in range(n_sim):
        prize = np.random.randint(3)         # Auto hinter einer Tür
        choice = np.random.randint(3)        # Ursprüngliche Wahl
        if switch:
            # Moderator öffnet eine Tür ohne Auto
            remaining = [i for i in range(3) if i != choice and i != prize]
            opened = np.random.choice(remaining)
            # Wechsel zur verbleibenden Tür
            choice = next(i for i in range(3) if i != choice and i != opened)
        if choice == prize:
            wins += 1
    return wins / n_sim

def monty_hall_widget(n_sim=10000, switch=True, seed=42):
    p_win = monty_hall_sim(n_sim=n_sim, switch=switch, seed=seed)
    strategy = "Wechseln" if switch else "Bleiben"
    print(f"Gewinnwahrscheinlichkeit bei Strategie '{strategy}': {p_win:.3f}")

    # Balkendiagramm mit Gewinn / Verloren
    plt.figure(figsize=(6,1))
    plt.bar(["Gewinn", "Verloren"], [p_win, 1-p_win], color=["orange","lightblue"])
    plt.ylim(0,1)
    plt.ylabel("Wahrscheinlichkeit")
    plt.title(f"{strategy} – {n_sim} Simulationen, Seed={seed}")
    plt.show()

interact(
    monty_hall_widget,
    n_sim=IntText(value=10000, description="Simulationen"),
    switch=Checkbox(value=True, description="Wechseln"),
    seed=IntText(value=42, description="Seed")
)
````
