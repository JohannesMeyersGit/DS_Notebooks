## 1. Motivation und Zusammenhang

Hypothesentests sind ein zentrales Werkzeug der **inferenzstatistischen Analyse**.  
Sie dienen dazu, anhand von Stichprobendaten **Aussagen über die Grundgesamtheit** zu treffen und zu prüfen, ob ein beobachteter Effekt **statistisch signifikant** ist oder durch **Zufall** erklärt werden kann.

Im Kontext von **Machine Learning (ML)** treten Hypothesentests überall dort auf, wo Entscheidungen auf Basis von Daten getroffen werden, z. B.:

- Ist **Modell A** wirklich besser als **Modell B** oder nur zufällig?  
- Führt ein **neues Feature** zu einer signifikanten Leistungsverbesserung?  
- Haben sich **Datenverteilungen** im Zeitverlauf verändert (Data/Concept Drift)?  
- Ist ein **Parameter oder Koeffizient** in einem linearen Modell signifikant ungleich $0$?

Der Zusammenhang zu **Konfidenzintervallen (CIs)** ist eng:  
- CIs liefern den Bereich plausibler Parameterwerte.  
- Hypothesentests prüfen gezielt, ob ein bestimmter Wert (z. B. Mittelwertdifferenz $=0$) **innerhalb oder außerhalb** dieses Bereichs liegt.

**Beispiel:**  
Wenn das 95 %-CI für die Genauigkeitsdifferenz zweier Modelle
$$
[-0.02,\; 0.08]
$$
ist, kann man $H_0:$ „Beide Modelle sind gleich gut“ **nicht verwerfen**, weil $0$ im Intervall liegt.

---

## 2. Grundlagen der Hypothesenbildung

Ein Hypothesentest vergleicht zwei konkurrierende Aussagen über die Population:

- **Nullhypothese ($H_0$)** — Ausgangsannahme: kein Effekt, kein Unterschied.  
  Beispiel: $H_0$: „Das neue Modell hat die gleiche Genauigkeit wie das alte.“

- **Alternativhypothese ($H_1$)** — Behauptung, die wir prüfen wollen: es existiert ein Effekt oder Unterschied.  
  Beispiel: $H_1$: „Das neue Modell hat eine andere (z. B. bessere) Genauigkeit.“

### Einseitige und zweiseitige Hypothesen

| Typ | Formulierung | Beispiel (Mittelwertvergleich) | Verwendung |
|-----|--------------:|:-------------------------------|:----------|
| **Zweiseitig** | $H_0: \mu_1 = \mu_2$  /  $H_1: \mu_1 \ne \mu_2$ | Unterschied in **beiden** Richtungen | Standardfall |
| **Einseitig (rechts)** | $H_0: \mu_1 \le \mu_2$  /  $H_1: \mu_1 > \mu_2$ | Prüfen auf Verbesserung | Leistungssteigerung |
| **Einseitig (links)** | $H_0: \mu_1 \ge \mu_2$  /  $H_1: \mu_1 < \mu_2$ | Prüfen auf Verschlechterung | Qualitätskontrolle |

---

## 3. Teststatistik und Entscheidungsregel

### Was ist eine Teststatistik?

Eine **Teststatistik** ist eine Funktion der Stichprobendaten, die in einer einzigen Zahl $T_{\text{emp}}$ (empirischer Testwert) zusammenfasst, wie stark die beobachteten Daten von der Nullhypothese abweichen:

$$
T_{\text{emp}} = f(X_1, X_2, \dots, X_n)
$$

Die Verteilung von $T$ unter $H_0$ ist bekannt (oder angenähert), und darauf basieren die Ableitungsregel (kritischer Bereich) und der p-Wert.

---

### Intuition

- $T_{\text{emp}}$ fasst die Evidenz gegen $H_0$ in einer Zahl zusammen.  
- Je stärker $T_{\text{emp}}$ von dem unter $H_0$ erwarteten Wert abweicht, desto eher verwerfen wir $H_0$.  
- Aus $T_{\text{emp}}$ lassen sich p-Wert und kritische Entscheidung direkt berechnen.

---

### Häufige Teststatistiken (Übersicht)

| Testart | Parameter | Teststatistik | Verteilung unter $H_0$ |
|---------|:---------:|:-------------:|:----------------------:|
| Z-Test | Mittelwert (bekanntes $\sigma$) | $T = \dfrac{\bar{x}-\mu_0}{\sigma/\sqrt{n}}$ | $N(0,1)$ |
| t-Test | Mittelwert (unbekanntes $\sigma$) | $T = \dfrac{\bar{x}-\mu_0}{s/\sqrt{n}}$ | $t$-Verteilung |
| Chi²-Test | Häufigkeiten / Varianz | $T = \displaystyle\sum_i \frac{(O_i-E_i)^2}{E_i}$ | $\chi^2$-Verteilung |
| F-Test / ANOVA | Verhältnis von Varianzen | $F = \dfrac{\text{MS}_B}{\text{MS}_W}$ | $F$-Verteilung |

---

### Entscheidungsregel

Vergleich von $T_{\text{emp}}$ mit dem kritischen Wert $T_{\text{kritisch}}(\alpha)$ (abhängig vom Signifikanzniveau $\alpha$):

- **Zweiseitiger Test**: verwerfe $H_0$ wenn
  $$
  |T_{\text{emp}}| > T_{\text{kritisch}}(1-\tfrac{\alpha}{2})
  $$
- **Rechtsseitiger Test**: verwerfe $H_0$ wenn $T_{\text{emp}} > T_{\text{kritisch}}(1-\alpha)$  
- **Linksseitiger Test**: verwerfe $H_0$ wenn $T_{\text{emp}} < T_{\text{kritisch}}(\alpha)$

Der kritische Bereich enthält diejenigen Werte, die unter $H_0$ sehr unwahrscheinlich sind (Wahrscheinlichkeit $<\alpha$).

---

### Beispiel: Mittelwertvergleich

Prüfung, ob Stichprobenmittel $\bar{x}=52$ mit $s=10$, $n=25$ von $\mu_0=50$ abweicht:

$$
T_{\text{emp}} = \frac{52-50}{10/\sqrt{25}} = 1.0
$$

Für $\alpha=0.05$ und $df=24$ gilt $T_{\text{kritisch}}\approx 2.06$. Da $1.0<2.06$, wird $H_0$ **nicht** verworfen.

---

### Kritischer Bereich vs. Akzeptanzbereich

| Bereich | Bedeutung |
|---------|----------|
| Kritischer Bereich | Werte, bei denen $H_0$ verworfen wird (Ereignisse mit $P<\alpha$ unter $H_0$) |
| Akzeptanzbereich | Werte, die mit $H_0$ vereinbar sind |

---

### Bedeutung in Machine Learning

- **Modellvergleich:** Teststatistik ist typischerweise die Differenz von Metriken (z. B. Accuracy) über Resamples/Folds.  
- **Feature-Relevanz:** t-Statistik für Koeffizienten $\hat\beta_j$:
  $$
  T_j = \frac{\hat\beta_j}{\mathrm{SE}(\hat\beta_j)}
  $$
  hoher $|T_j|$ → starke Evidenz, dass Feature $j$ beiträgt.  
- **Varianztests:** Vergleich der Fehlerstreuung zweier Modelle mit F-Test.

---

### Praktische Hinweise

- Prüfe Annahmen (Normalität, Unabhängigkeit, Homoskedastizität).  
- Bei Annahmeverletzungen: nichtparametrische Tests (Mann–Whitney, Wilcoxon) oder Bootstrapping verwenden.  
- Berichte immer: Punktwert, Effektgröße und CI zusätzlich zum p-Wert.

## 4. Fehlerarten und Teststärke

Hypothesentests tragen Risiken für Fehlentscheidungen:

| Fehlerart | Bedeutung | Symbol |
|-----------|----------:|:------:|
| Fehler 1. Art | $H_0$ fälschlich verwerfen (falscher Alarm) | $\alpha$ |
| Fehler 2. Art | $H_0$ fälschlich beibehalten (fehlende Entdeckung) | $\beta$ |

Die **Power** ist:
$$
\text{Power} = 1-\beta
$$
→ Wahrscheinlichkeit, einen echten Effekt zu entdecken.

---

### Einflussgrößen der Power

| Einfluss | Wirkung |
|---------:|:-------|
| Stichprobengröße $n$ | Größeres $n$ → höhere Power |
| Effektgröße $\Delta=$|$\mu_1-\mu_2|$ | Größeres $\Delta$ → höhere Power |
| Signifikanzniveau $\alpha$ | Größeres $\alpha$ → höhere Power (mehr false positives) |
| Streuung $\sigma$ | Größere $\sigma$ → geringere Power |

---

### Intuition (Visualisierung)

- Nullverteilung (unter $H_0$) vs. Alternativverteilung (unter $H_1$).  
- Weniger Überlappung → kleinere $\beta$ → größere Power.  
- Praktisch: größere Testsets oder größere Effekte machen Unterschiede leichter nachweisbar.

---

### Beispiel (ML)

Zwei Modelle: SVM 78 % vs. RF 81 % bei $n=50$ → geringe Power (Unterschied kann zufällig sein).  
Bei $n=1000$ wäre derselbe Unterschied sehr wahrscheinlich signifikant.

---

**Zusammenfassung (Abschnitte 1–3):**  
- $H_0/H_1$ formulieren, passende Teststatistik wählen, $T_{\text{emp}}$ mit $T_{\text{kritisch}}$ vergleichen.  
- Prüfe Annahmen, berichte p-Werte zusammen mit Effektgrößen und CIs, und beachte Power/Probeplanung.


```python
# --- Interaktive Visualisierung: Hypothesentest mit α, β und Power ---
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider
from scipy.stats import norm

def plot_test_stat(alpha=0.05, mu0=0, mu1=1, sigma=1):
    x = np.linspace(-4, 6, 500)
    null_dist = norm.pdf(x, mu0, sigma)
    alt_dist = norm.pdf(x, mu1, sigma)
    
    # Kritische Werte (zweiseitig)
    z_crit = norm.ppf(1 - alpha/2)
    crit_right = mu0 + z_crit * sigma
    crit_left = mu0 - z_crit * sigma
    
    # Fehler 2. Art β und Power
    beta = norm.cdf(crit_right, mu1, sigma) - norm.cdf(crit_left, mu1, sigma)
    power = 1 - beta
    
    plt.figure(figsize=(8, 4))
    plt.plot(x, null_dist, label="Nullverteilung $H_0$", lw=2)
    plt.plot(x, alt_dist, label="Alternativverteilung $H_1$", lw=2, ls='--')
    
    # Kritische Bereiche (rot)
    plt.fill_between(x, 0, null_dist, where=(x > crit_right) | (x < crit_left),
                     color='red', alpha=0.25, label=f"Kritischer Bereich (α={alpha:.3f})")
    # β-Bereich (grün)
    plt.fill_between(x, 0, alt_dist, where=(x < crit_right) & (x > crit_left),
                     color='green', alpha=0.25, label=f"Fehler 2. Art (β={beta:.3f})")
    
    # Vertikale Linien für Erwartungswerte
    plt.axvline(mu0, color='black', ls=':', label='$\mu_0$')
    plt.axvline(mu1, color='black', ls=':', label='$\mu_1$')
    
    # Power-Text
    plt.text(x.max()-1.5, max(alt_dist)*0.9, f"Power = {power:.3f}", color='blue', fontsize=12)
    
    plt.title("Hypothesentest – Null- und Alternativverteilung")
    plt.xlabel("Teststatistik T")
    plt.ylabel("Dichte")
    plt.legend()
    plt.grid(alpha=0.3)
    plt.show()

interact(
    plot_test_stat,
    alpha=FloatSlider(value=0.05, min=0.001, max=0.2, step=0.005, description="α"),
    mu0=FloatSlider(value=0, min=-2, max=2, step=0.1, description="μ₀"),
    mu1=FloatSlider(value=1, min=-2, max=3, step=0.1, description="μ₁"),
    sigma=FloatSlider(value=1, min=0.2, max=2, step=0.1, description="σ")
)
```

🎯 **Interaktive Visualisierung: Hypothesentest, Verteilungen und Fehlerarten**

Diese interaktive Grafik zeigt die Grundidee eines Hypothesentests anhand der
Nullverteilung ($H_0$) und der Alternativverteilung ($H_1$) einer Teststatistik.

---

### 🔍 Elemente der Visualisierung

- **Blaue Kurve:** Dichtefunktion der Nullverteilung $H_0$ — beschreibt, welche Werte der Teststatistik $T$ zu erwarten sind, wenn die Nullhypothese gilt (kein Effekt).  
- **Orange gestrichelte Kurve:** Dichtefunktion der Alternativverteilung $H_1$ — beschreibt, wie die Teststatistik aussieht, wenn tatsächlich ein Effekt existiert.  
- **Rote Flächen:** Kritischer Bereich. Werte der Teststatistik, die hier fallen, führen zur Verwerfung von $H_0$. Der Anteil entspricht dem Signifikanzniveau $\alpha$ – Wahrscheinlichkeit für einen Fehler 1. Art (falscher Alarm).  
- **Grüne Fläche:** Bereich unter $H_1$, der nicht im kritischen Bereich liegt. Zeigt die Wahrscheinlichkeit eines Fehlers 2. Art $\beta$ – Wahrscheinlichkeit, $H_0$ beizubehalten, obwohl $H_1$ wahr ist.  
- **Power-Beschriftung ($1-\beta$):** Wahrscheinlichkeit, $H_0$ korrekt zu verwerfen, wenn $H_1$ gilt. Je größer die Power, desto zuverlässiger der Test.  
- **Vertikale Linien bei $\mu_0$ und $\mu_1$:** Erwartungswerte der Teststatistik unter $H_0$ und $H_1$. Der Abstand zeigt die Effektgröße – je größer, desto leichter kann der Test die beiden Fälle unterscheiden.  

---

### ⚙️ Interaktive Parameter

| Parameter | Bedeutung | Interpretation |
|-----------|-----------|----------------|
| α (Alpha) | Signifikanzniveau | Bestimmt, wie „streng“ der Test ist. Kleineres $\alpha$ → geringeres Risiko für Fehlalarm, aber höhere Wahrscheinlichkeit, echte Effekte zu übersehen. |
| μ₀, μ₁   | Erwartungswerte der Null- und Alternativverteilung | Steuern den Effektunterschied. Größerer Abstand → geringere Überlappung → höhere Teststärke (Power). |
| σ (Sigma)| Standardabweichung | Beeinflusst die Streuung der Verteilungen. Größere $\sigma$ → stärkere Überlappung → geringere Power. |

---

### 💡 Interpretation

- Starke Überlappung von $H_0$ und $H_1$ → schwacher Test: hohe Unsicherheit, kleine Power.  
- Klare Trennung → starker Test: $H_0$ kann mit hoher Sicherheit verworfen werden.  
- Die interaktive Darstellung zeigt, wie Stichprobengröße, Effektgröße und $\alpha$ gemeinsam Teststärke, Fehlerwahrscheinlichkeiten und Power beeinflussen.

---

### 🔢 Mathematische Herleitung von α, β und Power

- **Fehler 1. Art ($\alpha$):** Wahrscheinlichkeit, $H_0$ fälschlicherweise zu verwerfen.

$$
\alpha = P(T \in K \mid H_0)
$$

- **Fehler 2. Art ($\beta$):** Wahrscheinlichkeit, $H_0$ nicht zu verwerfen, obwohl $H_1$ gilt.

$$
\beta = P(T \in \text{Nicht-Kritischer Bereich} \mid H_1)
= P(c_\text{left} \le T \le c_\text{right} \mid T \sim \mathcal{N}(\mu_1, \sigma^2))
$$

- **Teststärke (Power):**

$$
1-\beta = P(T \in K \mid H_1)
$$

---

### 📘 Zusammenhang mit Machine Learning

- $H_0$ entspricht oft der Annahme „Modelle sind gleich gut“.  
- Der Test prüft, ob ein beobachteter Unterschied (z. B. 2 % höhere Accuracy) statistisch signifikant ist.  
- Kleines $\alpha$ schützt vor Fehlalarmen, während großer Effekt $|\mu_1 - \mu_0|$ die Power erhöht.  
- Mit der Visualisierung sieht man sofort, wie sich $\alpha$, $\beta$ und Power beim Ändern der Parameter verhalten.


## 4. p-Wert, Signifikanzniveau und Entscheidungsregel

---

### 4.1 Definition des p-Werts

Der **p-Wert** ist die zentrale Entscheidungsgröße im Hypothesentest:

> Der p-Wert ist die Wahrscheinlichkeit, **unter der Annahme der Nullhypothese**, ein Ergebnis zu erhalten, das **mindestens so extrem** ist wie das beobachtete.

Formal:
$$
p = P(T \ge T_{\text{emp}} \mid H_0)
$$
für einen rechtsseitigen Test (bei links- oder zweiseitigen Tests analog angepasst).

---

### 4.2 Entscheidungsregel

Ein Ergebnis wird als **signifikant** bezeichnet, wenn:
$$
p < \alpha
$$

wobei  
- $\alpha$ das **Signifikanzniveau** ist (typisch 0.05, 0.01 oder 0.001)  
- $\alpha$ entspricht der **Fehlerwahrscheinlichkeit 1. Art**

| Signifikanzniveau | Bedeutung | Interpretation |
|--------------------|------------|----------------|
| $\alpha = 0.05$ | 5 % Risiko, $H_0$ fälschlich zu verwerfen | häufige Wahl |
| $\alpha = 0.01$ | 1 % Risiko | strenger Test |
| $\alpha = 0.001$ | sehr konservativ | z. B. in High-Stakes-Domains |

---

### 4.3 p-Wert im Kontext von Machine Learning

Im ML wird der p-Wert häufig genutzt, um:
- **Modellvergleiche** zu validieren (z. B. Accuracy-Unterschiede)
- **Feature-Signifikanz** in linearen / logistischen Modellen zu bewerten
- **A/B-Tests** zur Bewertung neuer Modelle, Algorithmen oder Parameter zu steuern

**Beispiel:**
> Wir vergleichen zwei Modelle auf demselben Testset.  
> Der beobachtete Unterschied beträgt 4 %.  
> Ein t-Test ergibt $p = 0.03$ →  
> Bei $\alpha = 0.05$ ist der Unterschied **signifikant**, d. h. mit 95 % Sicherheit nicht zufällig.

---

### 4.4 Interpretation und typische Fehlannahmen

| Falsch | Richtig |
|--------|----------|
| "Ein kleiner p-Wert beweist $H_1$." | p-Wert misst nur die *Kompatibilität* der Daten mit $H_0$. |
| "p = 0.05 heißt, es gibt 5 % Wahrscheinlichkeit, dass $H_0$ wahr ist." | Nein — $H_0$ ist entweder wahr oder falsch; der p-Wert bezieht sich auf die Daten, nicht auf $H_0$. |
| "Nicht-signifikant → kein Effekt." | Nicht-signifikant bedeutet: *Daten liefern keine ausreichende Evidenz*, nicht dass kein Effekt existiert. |

---

### 4.5 Verbindung zu Konfidenzintervallen

Beide Ansätze sind äquivalent:
- Wenn der Wert unter $H_0$ (z. B. $\Delta = 0$) **im CI liegt** → $p > \alpha$ → $H_0$ wird **nicht verworfen**
- Wenn er **außerhalb des CI liegt** → $p < \alpha$ → $H_0$ wird **verworfen**

---

### 4.6 Praktische Hinweise (ML-Kontext)

- **p-Werte sind empfindlich gegenüber Stichprobengröße:**  
  Bei großen Testsets wird fast jeder kleine Unterschied signifikant.
- **Immer Effektgröße und CI mitberichten!**
- **p-Werte ≠ praktische Relevanz:**  
  Eine signifikante Verbesserung von 0.1 % Accuracy kann praktisch bedeutungslos sein.

## 5. Klassische Hypothesentests

---

### 5.1 Überblick

| Test | Zweck | Verteilung | Anwendung |
|------|-------|-------------|-----------|
| **Z-Test** | Vergleich von Mittelwerten bei bekanntem $\sigma$ | Normalverteilung | Große $n$, bekannte Varianz |
| **t-Test** | Vergleich von Mittelwerten bei unbekanntem $\sigma$ | t-Verteilung | Kleine $n$, unbekannte Varianz |
| **Chi²-Test** | Test auf Unabhängigkeit oder Varianzvergleich | Chi²-Verteilung | Kategoriale Daten |
| **ANOVA (F-Test)** | Vergleich von mehr als zwei Gruppen | F-Verteilung | Mehrere Modelle oder Bedingungen |

---

### 5.2 t-Test (Student-Test)

#### (a) Einstichproben-t-Test
Prüft, ob der Mittelwert einer Stichprobe $\bar{x}$ von einem Sollwert $\mu_0$ abweicht:
$$
t = \frac{\bar{x} - \mu_0}{s / \sqrt{n}}
$$
mit $s$ = Stichprobenstandardabweichung.

#### (b) Zweistichproben-t-Test
Vergleicht zwei unabhängige Mittelwerte:
$$
t = \frac{\bar{x}_1 - \bar{x}_2}{\sqrt{s_p^2 \left(\frac{1}{n_1} + \frac{1}{n_2}\right)}}
$$
mit gepoolter Varianz:
$$
s_p^2 = \frac{(n_1-1)s_1^2 + (n_2-1)s_2^2}{n_1 + n_2 - 2}
$$

#### (c) Gepaarter t-Test
Für gepaarte Beobachtungen (z. B. A/B-Test auf demselben Testset):
$$
t = \frac{\bar{d}}{s_d / \sqrt{n}}, \quad d_i = x_{1i} - x_{2i}
$$

---

### 5.3 Anwendung: Modellvergleich im ML

**Beispiel:**
Wir testen, ob Modell A (SVM) und Modell B (Random Forest) sich signifikant unterscheiden.

1. Berechne Genauigkeiten beider Modelle pro Bootstrap-Sample oder Cross-Validation-Fold.  
2. Erhalte Differenzen $d_i = \text{acc}_A^{(i)} - \text{acc}_B^{(i)}$.  
3. Führe gepaarten t-Test durch:
   $$
   t = \frac{\bar{d}}{s_d / \sqrt{n}}
   $$
4. Berechne p-Wert und vergleiche mit $\alpha$.

**Interpretation:**
- $p < 0.05$ → Modelle unterscheiden sich signifikant.  
- $p \ge 0.05$ → keine Evidenz für Unterschied (möglicherweise gleiche Leistung).

---

### 5.4 Chi²-Test (z. B. für Feature Relevance)

Der Chi²-Test misst die Abweichung beobachteter Häufigkeiten $O_i$ von erwarteten Häufigkeiten $E_i$:
$$
\chi^2 = \sum_i \frac{(O_i - E_i)^2}{E_i}
$$

Anwendung im ML:
- Prüfung, ob ein Feature **unabhängig vom Ziel** ist → Feature-Selektion
- Prüfung, ob **Modellvorhersagen von Klassenverteilungen** abweichen

---

### 5.5 ANOVA (Analysis of Variance)

Vergleicht **mehr als zwei Gruppen oder Modelle** auf Mittelwertunterschiede.

Teststatistik:
$$
F = \frac{\text{MS}_{\text{between}}}{\text{MS}_{\text{within}}}
$$
mit:
- $\text{MS}_{\text{between}} =$ mittlere Varianz **zwischen** den Gruppen
- $\text{MS}_{\text{within}} =$ mittlere Varianz **innerhalb** der Gruppen

**Beispiel (ML):**
Vergleich von 3 Modellen auf denselben Daten.  
Ein signifikanter F-Wert ($p < \alpha$) → mindestens ein Modell unterscheidet sich signifikant.

→ Post-hoc-Tests (z. B. Tukey-HSD) zeigen, *welches* Modell sich unterscheidet.

---

### 5.6 Zusammenfassung der klassischen Tests

| Test | Hypothese | Typische Anwendung im ML | Annahmen |
|------|-------------|--------------------------|-----------|
| **t-Test** | Mittelwertgleichheit | Modellvergleich, Feature-Impact | Normalverteilung, Varianzgleichheit |
| **Chi²-Test** | Unabhängigkeit | Feature-Selektion, Diskrete Targets | Erwartete Häufigkeiten > 5 |
| **ANOVA** | Mehrgruppenvergleich | Vergleich mehrerer Modelle / Features | Normalverteilung, Varianzgleichheit |
| **Z-Test** | Mittelwertvergleich (bekanntes σ) | Große Stichproben | CLT-Gültigkeit |

---

📘 **Kernaussage:**
> Hypothesentests sind das Fundament, um Unterschiede, Korrelationen und Effekte in ML-Modellen **statistisch abzusichern**, anstatt sie nur auf beobachtete Punktwerte zu stützen.

---


##  A/B-Test: Grundlagen und Anwendungen

Ein **A/B-Test** (oder Split-Test) ist eine standardisierte Methode, um zwei Varianten eines Systems, Produkts oder Modells zu vergleichen und zu prüfen, ob sich ein beobachteter Unterschied statistisch signifikant zeigt.

---

### 1 Hintergrund

- **Ziel:** Prüfen, ob zwei Gruppen (A und B) unterschiedlich reagieren.
- **Nullhypothese ($H_0$):** Es gibt keinen Unterschied zwischen den Gruppen.
- **Alternativhypothese ($H_1$):** Es gibt einen Unterschied zwischen den Gruppen.
- **Datenbasis:** Häufig numerische oder binäre Messwerte (z. B. Conversion Rate, Klickrate, Accuracy).

**Typische Schritte:**
1. Definition der Zielgröße (z. B. Conversion Rate oder Modellgenauigkeit)
2. Aufteilung in zwei Gruppen (A und B)
3. Sammlung von Messwerten
4. Durchführung eines statistischen Tests (z. B. t-Test, z-Test)
5. Interpretation von p-Wert, Signifikanzniveau α und Teststärke (Power)

---

### 2 Anwendungsbeispiele

**Marketing / Webanalyse:**
- Vergleich zweier Webseiten-Versionen  
- Optimierung der Conversion Rate  
- Testen von Werbekampagnen

**Machine Learning:**
- Vergleich zweier Modelle oder Feature-Sets  
- Prüfen, ob die Accuracy, F1-Score oder andere Metriken signifikant unterschiedlich sind  
- Sicherstellen, dass Verbesserungen nicht durch Zufall entstehen

**Produkt- und UX-Design:**
- Testen neuer Funktionen oder Layouts  
- Validieren von Designänderungen auf Nutzerinteraktion

---

### 3 Zusammenhang zu Hypothesentests

- A/B-Test ist ein konkretes Beispiel für einen **zweiseitigen Hypothesentest**.
- **Signifikanzniveau ($\alpha$):** Risiko, fälschlicherweise einen Unterschied zu sehen.
- **Fehler 2. Art ($\beta$):** Risiko, einen echten Unterschied zu übersehen.
- **Power ($1-\beta$):** Wahrscheinlichkeit, einen echten Unterschied korrekt zu erkennen.

---

 **Merke:**  
A/B-Tests verbinden die praktische Fragestellung („Welche Variante ist besser?“) mit formaler Statistik und helfen, Entscheidungen datenbasiert zu treffen – sei es in Webanalyse, Marketing oder Machine Learning.


## 🧪 Didaktisches Beispiel: A/B-Test auf einer Webseite

**Situation:**  
Eine Firma möchte testen, welche von zwei Webseiten-Versionen (A oder B) mehr Besucher dazu bringt, sich anzumelden (Conversion).

- **Gruppe A:** Alte Version der Webseite  
- **Gruppe B:** Neue Version der Webseite  

**Ziel:** Herausfinden, ob die neue Version B die Conversion Rate tatsächlich verbessert.

---

### 1️⃣ Datensimulation

- 50 Besucher pro Version  
- Conversion = 1 (erfolgreich) oder 0 (kein Erfolg)  

Wir simulieren:

- Gruppe A: 10 % Conversion  
- Gruppe B: 15 % Conversion  

---

### 2️⃣ Hypothesen

- **Nullhypothese ($H_0$):** Conversion A = Conversion B (kein Unterschied)  
- **Alternativhypothese ($H_1$):** Conversion B > Conversion A (B ist besser)  

---

### 3️⃣ Durchführung des Tests

- Wir benutzen einen **t-Test** für unabhängige Stichproben  
- Berechnung von **t-Statistik** und **p-Wert**  
- Vergleich des p-Werts mit Signifikanzniveau $\alpha$ (z. B. 0.05)  

**Interpretation:**  

- $p < 0.05$ → Unterschied ist statistisch signifikant → $H_0$ ablehnen  
- $p \ge 0.05$ → Kein signifikanter Unterschied → $H_0$ beibehalten  

---

### 4️⃣ Visualisierung

- Histogramme der beiden Gruppen  
- Überlappung der Balken zeigt Unsicherheit  
- Wenn B mehr "1" hat als A → höhere Chance, $H_0$ zu verwerfen  

---

### 5️⃣ Verbindung zu Hypothesentest-Konzepten

- **Rote Fläche:** Wahrscheinlichkeit α, fälschlich $H_0$ abzulehnen  
- **Grüne Fläche:** Wahrscheinlichkeit β, echten Unterschied zu übersehen  
- **Power ($1-\beta$):** Wahrscheinlichkeit, einen echten Unterschied zu erkennen  

💡 **Merke:**  
Dieses einfache Beispiel zeigt, wie ein A/B-Test **praktisch die Hypothesenprüfung** umsetzt: Daten sammeln, testen, interpretieren, Entscheidung treffen.


```python
# --- Interaktives A/B-Test Widget mit Balken der beobachteten Daten ---
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, IntSlider, FloatSlider
from scipy.stats import ttest_ind, norm

def ab_test_widget(n_a=50, n_b=50, p_a=0.1, p_b=0.15, alpha=0.05):
    np.random.seed(42)
    
    # Daten simulieren (0 = kein Erfolg, 1 = Erfolg)
    group_a = np.random.binomial(1, p_a, n_a)
    group_b = np.random.binomial(1, p_b, n_b)
    
    # t-Test
    t_stat, p_value = ttest_ind(group_a, group_b, equal_var=False)
    
    # Schätzung der Power
    mean_a, mean_b = group_a.mean(), group_b.mean()
    std_a, std_b = np.sqrt(p_a*(1-p_a)/n_a), np.sqrt(p_b*(1-p_b)/n_b)
    se_diff = np.sqrt(std_a**2 + std_b**2)
    
    z_crit = norm.ppf(1 - alpha/2)
    crit_left = mean_a - z_crit * se_diff
    crit_right = mean_a + z_crit * se_diff
    
    beta = norm.cdf(crit_right, loc=mean_b, scale=se_diff) - norm.cdf(crit_left, loc=mean_b, scale=se_diff)
    power = 1 - beta
    
    # Dichtekurven
    x = np.linspace(-0.1, 1.1, 200)
    dist_a = norm.pdf(x, mean_a, std_a)
    dist_b = norm.pdf(x, mean_b, std_b)
    
    plt.figure(figsize=(8, 5))
    
    # Dichtekurven
    plt.plot(x, dist_a, label="Gruppe A (Nullhypothese)", lw=2)
    plt.plot(x, dist_b, label="Gruppe B (Alternativhypothese)", lw=2, ls='--')
    
    # Kritische Bereiche
    plt.fill_between(x, 0, dist_a, where=(x < crit_left) | (x > crit_right), color='red', alpha=0.25, label=f"α={alpha:.3f}")
    plt.fill_between(x, 0, dist_b, where=(x > crit_left) & (x < crit_right), color='green', alpha=0.25, label=f"β={beta:.3f}")
    
    # Vertikale Linien
    plt.axvline(mean_a, color='black', ls=':', label='Mittelwert A')
    plt.axvline(mean_b, color='black', ls=':', label='Mittelwert B')
    
    # Power-Text
    plt.text(0.65, max(dist_b)*0.9, f"Power = {power:.3f}", color='blue', fontsize=12)
    
    # Balken für die beobachteten Daten
    counts_a = [np.sum(group_a==0), np.sum(group_a==1)]
    counts_b = [np.sum(group_b==0), np.sum(group_b==1)]
    plt.bar([-0.05,0.95], counts_a/np.sum(counts_a)*max(dist_b), width=0.1, alpha=0.4, color='blue', label='Beobachtete A')
    plt.bar([0.05,1.05], counts_b/np.sum(counts_b)*max(dist_b), width=0.1, alpha=0.4, color='orange', label='Beobachtete B')
    
    plt.title(f"A/B-Test Simulation\n t={t_stat:.2f}, p={p_value:.3f}")
    plt.xlabel("Erfolg (0 = kein Erfolg, 1 = Erfolg)")
    plt.ylabel("Dichte")
    plt.xticks([0,1])
    plt.legend()
    plt.grid(alpha=0.3)
    plt.show()

interact(
    ab_test_widget,
    n_a=IntSlider(value=50, min=10, max=200, step=10, description="n A"),
    n_b=IntSlider(value=50, min=10, max=200, step=10, description="n B"),
    p_a=FloatSlider(value=0.1, min=0.01, max=0.5, step=0.01, description="p A"),
    p_b=FloatSlider(value=0.15, min=0.01, max=0.5, step=0.01, description="p B"),
    alpha=FloatSlider(value=0.05, min=0.01, max=0.2, step=0.01, description="α")
)
```
