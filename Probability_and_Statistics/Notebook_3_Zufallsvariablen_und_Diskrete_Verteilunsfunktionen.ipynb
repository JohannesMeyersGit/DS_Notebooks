## 1. Grundbegriffe: Zufallsvariablen

Eine **Zufallsvariable** $X$ ist eine Funktion, die jedem möglichen Ergebnis eines Zufallsexperiments eine Zahl zuordnet. Sie beschreibt also **quantitativ**, welche Ergebnisse eines Experiments eintreten können.

- **Diskrete Zufallsvariable:** Nimmt nur **abzählbar viele Werte** an (z. B. Augenzahl eines Würfels: 1,2,3,4,5,6)
- **Stetige Zufallsvariable:** Nimmt Werte aus einem **kontinuierlichen Intervall** an (z. B. Körpergröße in cm)

### Wahrscheinlichkeitsfunktion (PMF)

Für diskrete Zufallsvariablen beschreibt die **Wahrscheinlichkeitsfunktion (Probability Mass Function, PMF)** die Wahrscheinlichkeit für jedes mögliche Ergebnis:

$$
P(X=x_i) = \text{Wahrscheinlichkeit, dass } X \text{ den Wert } x_i \text{ annimmt.}
$$

**Eigenschaften:**
1. $0 \leq P(X=x_i) \leq 1$
2. $\sum_i P(X=x_i) = 1$

**Beispiel:** Ein fairer Würfel

| Augenzahl $x$ | 1 | 2 | 3 | 4 | 5 | 6 |
|---------------|---|---|---|---|---|---|
| $P(X=x)$      | 1/6 | 1/6 | 1/6 | 1/6 | 1/6 | 1/6 |

### Mini-Grafik

Eine einfache Visualisierung der PMF für einen Würfel zeigt die **Wahrscheinlichkeiten auf der y-Achse** und die **Augenzahlen auf der x-Achse**. Jede Säule stellt die Wahrscheinlichkeit eines Ergebnisses dar.

```python
import matplotlib.pyplot as plt
import numpy as np
from ipywidgets import interact, FloatSlider

labels = [1,2,3,4,5,6]
labels_text = ["Augenzahl 1", "Augenzahl 2", "Augenzahl 3", "Augenzahl 4", "Augenzahl 5", "Augenzahl 6"]

def plot_pmf(p1=1/6, p2=1/6, p3=1/6, p4=1/6, p5=1/6, p6=1/6):
    probs = np.array([p1, p2, p3, p4, p5, p6])
    probs = probs / probs.sum()  # Normierung
    
    plt.figure(figsize=(10,5))
    bars = plt.bar(labels, probs, color='skyblue', edgecolor='black')
    
    for bar, prob, text in zip(bars, probs, labels_text):
        plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.02, 
                 f"{text}\nP={prob:.2f}", ha='center', va='bottom', fontsize=10)
    
    plt.ylim(0, 1)
    plt.xticks(labels, labels_text, rotation=0)
    plt.ylabel("Wahrscheinlichkeit P(X)")
    plt.xlabel("Ergebnis der Zufallsvariablen X")
    plt.title("Wahrscheinlichkeitsverteilung einer diskreten Zufallsvariablen (PMF)")
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.show()

interact(plot_pmf,
         p1=FloatSlider(value=1/6, min=0, max=1, step=0.01, description='P(1)'),
         p2=FloatSlider(value=1/6, min=0, max=1, step=0.01, description='P(2)'),
         p3=FloatSlider(value=1/6, min=0, max=1, step=0.01, description='P(3)'),
         p4=FloatSlider(value=1/6, min=0, max=1, step=0.01, description='P(4)'),
         p5=FloatSlider(value=1/6, min=0, max=1, step=0.01, description='P(5)'),
         p6=FloatSlider(value=1/6, min=0, max=1, step=0.01, description='P(6)')
)
```

## 4.3 Erwartungswert, Varianz und Momente

### Erwartungswert (Mean)
Der **Erwartungswert** einer diskreten Zufallsvariablen $X$ ist das **gewichtete Mittel aller möglichen Werte**, wobei die Gewichte die Wahrscheinlichkeiten der jeweiligen Ausprägungen sind. Er gibt eine zentrale Tendenz oder den „Durchschnittswert“ an, den man bei vielen Wiederholungen des Zufallsexperiments erwarten würde.

Formel:
$$
\mathbb{E}[X] = \sum_{x} x \cdot p_X(x)
$$
- $x$: mögliche Ausprägungen der Zufallsvariablen $X$  
- $p_X(x)$: Wahrscheinlichkeit, dass $X = x$  

**Beispiel: fairer Würfel**
- Mögliche Werte: $1,2,3,4,5,6$  
- Wahrscheinlichkeiten: $p_X(x) = 1/6$ für alle $x$  

Berechnung:
$$
\mathbb{E}[X] = 1 \cdot \frac{1}{6} + 2 \cdot \frac{1}{6} + 3 \cdot \frac{1}{6} + 4 \cdot \frac{1}{6} + 5 \cdot \frac{1}{6} + 6 \cdot \frac{1}{6} = 3.5
$$

---

### Varianz (Variance)
Die **Varianz** misst, wie stark die Werte der Zufallsvariablen um den Erwartungswert streuen. Sie ist ein Maß für die **Streuung oder Unsicherheit** der Zufallsvariable.

Formel:
$$
\text{Var}(X) = \mathbb{E}[(X - \mathbb{E}[X])^2] = \sum_x (x - \mathbb{E}[X])^2 \cdot p_X(x)
$$

**Beispiel: Würfel**
- Erwartungswert: $\mathbb{E}[X] = 3.5$  
- Abweichungen: $1-3.5=-2.5$, $2-3.5=-1.5$, …  
- Quadrate der Abweichungen: $(-2.5)^2=6.25$, $(-1.5)^2=2.25$, …  
- Multiplikation mit den Wahrscheinlichkeiten $1/6$ und Summe:
$$
\text{Var}(X) = \frac{6.25 + 2.25 + 0.25 + 0.25 + 2.25 + 6.25}{6} = \frac{17.5}{6} \approx 2.917
$$

---

### Momente
**Momente** einer Zufallsvariablen geben Informationen über die Form der Verteilung über den Erwartungswert hinaus.  
Das $k$-te Moment ist definiert als:
$$
\mu_k = \mathbb{E}[X^k] = \sum_x x^k \cdot p_X(x)
$$

- **1. Moment:** Erwartungswert $\mathbb{E}[X]$  
- **2. zentriertes Moment:** Varianz $\text{Var}(X) = \mathbb{E}[(X-\mathbb{E}[X])^2]$  
- **3. zentriertes Moment:** misst die **Schiefe** (ob die Verteilung links- oder rechtslastig ist)  
- **4. zentriertes Moment:** misst die **Wölbung (Kurtosis)** der Verteilung  

**Beispiel Würfel:**
- 1. Moment: $\mu_1 = \mathbb{E}[X] = 3.5$  
- 2. zentriertes Moment: $\mu_2 = \text{Var}(X) \approx 2.917$  
- 3. zentriertes Moment (Schiefe): $\mu_3 = 0$ (symmetrisch)  
- 4. zentriertes Moment (Kurtosis): $\mu_4 = \sum (x-3.5)^4 \cdot \frac{1}{6} \approx 4.958$  

**Hinweis:** Momente helfen, die **Form der Verteilung** besser zu verstehen, nicht nur die zentrale Tendenz.


## 4.4 Diskrete Verteilungen

Diskrete Zufallsvariablen nehmen **abzählbar viele Werte** an. Die drei wichtigsten Verteilungen für Data Science sind **Bernoulli**, **Binomial** und **Poisson**. Im Folgenden werden sie mit ihren Wahrscheinlichkeiten, Erwartungswerten, Varianzen, Momenten und typischen Anwendungen dargestellt.

---

### 4.4.1 Bernoulli-Verteilung

- Beschreibung: Ein Experiment mit **zwei möglichen Ergebnissen**: Erfolg ($1$) oder Misserfolg ($0$).  
- Parameter: $p = P(X=1)$
- Wahrscheinlichkeitsfunktion (PMF):  
$$
P(X=x) = p^x (1-p)^{1-x}, \quad x \in \{0,1\}
$$
- Erwartungswert:  
$$
\mathbb{E}[X] = p
$$
- Varianz:  
$$
\text{Var}(X) = p(1-p)
$$
- $k$-tes Moment:  
$$
\mu_k = \mathbb{E}[X^k] = p
$$

**Anwendungen:**  
- Biologie: Erfolg/Nichterfolg bei Genexperimenten (z. B. Mutation auftritt oder nicht)  
- Physik: Zerfall eines einzelnen radioaktiven Atoms in einem Zeitintervall  
- Data Science / ML: Klassifikation von binären Labels, Feature Engineering für diskrete Variablen  

**Beispiel:** Münzwurf mit $P(\text{Kopf}) = 0.6$ → $\mathbb{E}[X] = 0.6$, $\text{Var}(X) = 0.24$

---

### 4.4.2 Binomial-Verteilung

- Beschreibung: Summe von $n$ **unabhängigen Bernoulli-Experimenten**.  
- Parameter: $n$ (Anzahl Versuche), $p$ (Erfolgswahrscheinlichkeit)
- PMF:  
$$
P(X=k) = \binom{n}{k} p^k (1-p)^{n-k}, \quad k = 0,...,n
$$
- Erwartungswert:  
$$
\mathbb{E}[X] = n p
$$
- Varianz:  
$$
\text{Var}(X) = n p (1-p)
$$
- $k$-tes Moment:  
$$
\mu_k = \mathbb{E}[X^k], \quad k=1,2,\dots
$$

**Anwendungen:**  
- Biologie: Anzahl erfolgreicher Überleben bei $n$ Individuen  
- Physik: Anzahl geladener Teilchen, die einen Detektor passieren  
- Data Science / ML: Modellierung von Anzahl positiver Events in festen Stichproben, z. B. Klicks auf Anzeigen  

**Beispiel:** 10 Würfe einer Münze mit $p=0.5$ → $\mathbb{E}[X] = 5$, $\text{Var}(X) = 2.5$

---

### 4.4.3 Poisson-Verteilung

- Beschreibung: Modelliert die **Anzahl von Ereignissen** in einem festen Zeit- oder Raumbereich.  
- Parameter: $\lambda$ = erwartete Anzahl Ereignisse
- PMF:  
$$
P(X=k) = \frac{\lambda^k e^{-\lambda}}{k!}, \quad k = 0,1,2,...
$$
- Erwartungswert:  
$$
\mathbb{E}[X] = \lambda
$$
- Varianz:  
$$
\text{Var}(X) = \lambda
$$
- $k$-tes Moment:  
$$
\mu_k = \mathbb{E}[X^k], \quad k=1,2,...
$$

**Anwendungen:**  
- Biologie: Anzahl Mutationen pro DNA-Strang in einem festen Abschnitt  
- Physik: Anzahl Teilchendetektionen pro Zeiteinheit (Radioaktivität)  
- Data Science / ML: Ereigniszählung in Zeitreihen, z. B. Anzahl Server-Anfragen, Fehlermeldungen oder Kundenanrufe  

**Beispiel:** Anzahl Anrufe in einem Callcenter pro Stunde, $\lambda = 4$ → $\mathbb{E}[X] = 4$, $\text{Var}(X) = 4$

---


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import bernoulli, binom, poisson
from ipywidgets import interact, IntSlider, FloatSlider

def plot_discrete_distributions(p=0.5, n=10, k_max=15, lam=3):
    fig, axs = plt.subplots(1,3, figsize=(18,4))
    
    # --- Bernoulli ---
    x_bern = [0,1]
    y_bern = bernoulli.pmf(x_bern, p)
    mean_bern = bernoulli.mean(p)
    var_bern = bernoulli.var(p)
    axs[0].bar(x_bern, y_bern, color='skyblue')
    axs[0].set_title(f'Bernoulli PMF (p={p})\nE[X]={mean_bern:.2f}, Var[X]={var_bern:.2f}')
    axs[0].set_xlabel('x')
    axs[0].set_ylabel('P(X=x)')
    for i, v in enumerate(y_bern):
        axs[0].text(x_bern[i], v+0.02, f"{v:.2f}", ha='center')

    # --- Binomial ---
    x_bin = np.arange(0, n+1)
    y_bin = binom.pmf(x_bin, n, p)
    mean_bin = binom.mean(n, p)
    var_bin = binom.var(n, p)
    axs[1].bar(x_bin, y_bin, color='lightgreen')
    axs[1].set_title(f'Binomial PMF (n={n}, p={p})\nE[X]={mean_bin:.2f}, Var[X]={var_bin:.2f}')
    axs[1].set_xlabel('x')
    axs[1].set_ylabel('P(X=x)')
    for i, v in enumerate(y_bin):
        axs[1].text(x_bin[i], v+0.01, f"{v:.2f}", ha='center')

    # --- Poisson ---
    x_pois = np.arange(0, k_max+1)
    y_pois = poisson.pmf(x_pois, lam)
    mean_pois = poisson.mean(lam)
    var_pois = poisson.var(lam)
    axs[2].bar(x_pois, y_pois, color='salmon')
    axs[2].set_title(f'Poisson PMF (λ={lam})\nE[X]={mean_pois:.2f}, Var[X]={var_pois:.2f}')
    axs[2].set_xlabel('x')
    axs[2].set_ylabel('P(X=x)')
    for i, v in enumerate(y_pois):
        axs[2].text(x_pois[i], v+0.01, f"{v:.2f}", ha='center')

    plt.tight_layout()
    plt.show()

# Interaktive Widgets
interact(
    plot_discrete_distributions,
    p=FloatSlider(value=0.5, min=0.0, max=1.0, step=0.05, description='p (Bernoulli/Binomial)'),
    n=IntSlider(value=10, min=1, max=20, step=1, description='n (Binomial)'),
    k_max=IntSlider(value=15, min=5, max=30, step=1, description='Max x (Poisson)'),
    lam=FloatSlider(value=3, min=0.1, max=10, step=0.5, description='λ (Poisson)')
)
```

## 5. Kontinuierliche Verteilungen

Kontinuierliche Zufallsvariablen können unendlich viele Werte in einem Intervall annehmen. Im Gegensatz zu diskreten Variablen arbeiten wir hier mit der **Wahrscheinlichkeitsdichtefunktion (PDF)**. Die Wahrscheinlichkeit, dass die Variable einen exakten Wert annimmt, ist 0; man betrachtet stattdessen Wahrscheinlichkeiten über Intervalle.

### 5.1 Normalverteilung (Gauß)
- Charakterisiert durch Mittelwert $\mu$ und Standardabweichung $\sigma$.  
- Dichtefunktion:
$$
f(x) = \frac{1}{\sigma \sqrt{2\pi}} \exp\Big(-\frac{(x-\mu)^2}{2\sigma^2}\Big)
$$
- Erwartungswert: $\mathbb{E}[X] = \mu$  
- Varianz: $\text{Var}(X) = \sigma^2$  
- Anwendungen: Naturwissenschaften, Messfehler, Finanzdaten, Machine Learning (z. B. Standardisierung, probabilistische Modelle).  
- Typische Visualisierung: Glockenkurve, symmetrisch um $\mu$.

### 5.2 Exponentialverteilung
- Modelliert die Wartezeit zwischen Ereignissen, die zufällig und unabhängig auftreten (z. B. Radioaktiver Zerfall, Service-Zeiten in Warteschlangen).  
- Parameter: $\lambda > 0$ (Rate)  
$$
f(x) = \lambda e^{-\lambda x}, \quad x \ge 0
$$
- Erwartungswert: $\mathbb{E}[X] = 1/\lambda$  
- Varianz: $\text{Var}(X) = 1/\lambda^2$  
- Anwendungen: Warteschlangenmodelle, Zuverlässigkeitstechnik, Survival Analysis.

### 5.3 Gamma-Verteilung
- Generalisierung der Exponentialverteilung; modelliert die Summe von $k$ unabhängigen Exponentialvariablen.  
- Parameter: Form $k>0$, Skala $\theta>0$  
$$
f(x) = \frac{1}{\Gamma(k) \theta^k} x^{k-1} e^{-x/\theta}, \quad x \ge 0
$$
- Erwartungswert: $\mathbb{E}[X] = k\theta$  
- Varianz: $\text{Var}(X) = k\theta^2$  
- Anwendungen: Biologie, Queuing-Theorie, Bayesian Inference (Gamma-Prior), Lebensdauer-Analysen.

### 5.4 Gleichverteilung (Uniform)
- Alle Werte im Intervall $[a,b]$ sind gleich wahrscheinlich.  
$$
f(x) = \frac{1}{b-a}, \quad x \in [a,b]
$$
- Erwartungswert: $\mathbb{E}[X] = \frac{a+b}{2}$  
- Varianz: $\text{Var}(X) = \frac{(b-a)^2}{12}$  
- Anwendungen: Zufallszahlengenerierung, Monte-Carlo-Simulationen, Sampling, Baseline-Modelle in Statistik und Machine Learning.


```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider, Dropdown

def plot_continuous_distribution(dist, mu=0, sigma=1, lam=1, k=2, theta=1, a=0, b=1):
    x = np.linspace(-5, 10, 500)
    
    plt.figure(figsize=(8,5))
    
    if dist == "Normal":
        y = 1/(sigma*np.sqrt(2*np.pi)) * np.exp(-(x-mu)**2/(2*sigma**2))
        plt.plot(x, y, label=f"Normal(μ={mu}, σ={sigma})")
        plt.axvline(mu, color='red', linestyle='--', label=f"μ = {mu}")
        plt.axvline(mu + sigma, color='green', linestyle='--', label=f"μ+σ = {mu+sigma}")
        plt.axvline(mu - sigma, color='green', linestyle='--', label=f"μ-σ = {mu-sigma}")
        
    elif dist == "Exponential":
        x = np.linspace(0, 10, 500)
        y = lam * np.exp(-lam*x)
        plt.plot(x, y, label=f"Exponential(λ={lam})")
        plt.axvline(1/lam, color='red', linestyle='--', label=f"E[X] = {1/lam}")
        plt.axvline(1/lam + 1/lam, color='green', linestyle='--', label=f"E[X]+σ = {2/lam}")
    
    elif dist == "Gamma":
        from scipy.stats import gamma
        x = np.linspace(0, 20, 500)
        y = gamma.pdf(x, a=k, scale=theta)
        plt.plot(x, y, label=f"Gamma(k={k}, θ={theta})")
        plt.axvline(k*theta, color='red', linestyle='--', label=f"E[X] = {k*theta}")
        plt.axvline(k*theta + np.sqrt(k*theta**2), color='green', linestyle='--', label=f"E[X]+σ = {k*theta + np.sqrt(k*theta**2)}")
    
    elif dist == "Uniform":
        x = np.linspace(a-1, b+1, 500)
        y = np.array([1/(b-a) if a <= xi <= b else 0 for xi in x])
        plt.plot(x, y, label=f"Uniform({a},{b})")
        plt.axvline((a+b)/2, color='red', linestyle='--', label=f"E[X] = {(a+b)/2}")
        plt.axvline(a, color='green', linestyle='--', label=f"a = {a}")
        plt.axvline(b, color='green', linestyle='--', label=f"b = {b}")
    
    plt.title(f"{dist}-Verteilung")
    plt.xlabel("x")
    plt.ylabel("PDF")
    plt.legend()
    plt.grid(True)
    plt.show()

interact(
    plot_continuous_distribution,
    dist=Dropdown(options=["Normal","Exponential","Gamma","Uniform"], value="Normal", description="Verteilung:"),
    mu=FloatSlider(min=-5, max=5, step=0.1, value=0, description="μ"),
    sigma=FloatSlider(min=0.1, max=5, step=0.1, value=1, description="σ"),
    lam=FloatSlider(min=0.1, max=5, step=0.1, value=1, description="λ"),
    k=FloatSlider(min=1, max=10, step=1, value=2, description="k"),
    theta=FloatSlider(min=0.1, max=5, step=0.1, value=1, description="θ"),
    a=FloatSlider(min=-5, max=5, step=0.1, value=0, description="a"),
    b=FloatSlider(min=-5, max=10, step=0.1, value=1, description="b")
)
```

## 6. Eigenschaften kontinuierlicher Verteilungen

Kontinuierliche Zufallsvariablen unterscheiden sich von diskreten darin, dass sie **unendlich viele Werte in einem Intervall annehmen** können. Die Eigenschaften dieser Verteilungen lassen sich durch Funktionen und Kennwerte beschreiben.

---

### 6.1 PDF – Probability Density Function

- Die **Wahrscheinlichkeitsdichtefunktion (PDF)** $f_X(x)$ beschreibt die Dichte der Wahrscheinlichkeit an jedem Punkt $x$.  
- Für ein Intervall $[a,b]$ gilt:
$$
P(a \le X \le b) = \int_a^b f_X(x) \, dx
$$
- **Wichtig:** $f_X(x)$ selbst ist keine Wahrscheinlichkeit, sondern eine Dichte. Die Fläche unter der Kurve über das gesamte Definitionsintervall ist 1:
$$
\int_{-\infty}^{\infty} f_X(x) \, dx = 1
$$

**Beispiel:** Normalverteilung $N(\mu, \sigma^2)$:
$$
f_X(x) = \frac{1}{\sqrt{2\pi\sigma^2}} \exp\Big(-\frac{(x-\mu)^2}{2\sigma^2}\Big)
$$

---

### 6.2 CDF – Cumulative Distribution Function

- Die **Verteilungsfunktion (CDF)** $F_X(x)$ gibt die Wahrscheinlichkeit an, dass $X$ einen Wert kleiner oder gleich $x$ annimmt:
$$
F_X(x) = P(X \le x) = \int_{-\infty}^{x} f_X(t) \, dt
$$
- Die CDF ist **monoton steigend** und liegt zwischen 0 und 1.
- **Beispiel:** Für die Standardnormalverteilung $N(0,1)$:
$$
F_X(x) = \frac{1}{2} \Big[1 + \text{erf}\Big(\frac{x}{\sqrt{2}}\Big)\Big]
$$
- Die CDF ist hilfreich, um **Wahrscheinlichkeiten über Intervalle** einfach zu berechnen:
$$
P(a \le X \le b) = F_X(b) - F_X(a)
$$

---

### 6.3 Symmetrie & Standardisierung

- Viele Verteilungen, insbesondere die **Normalverteilung**, sind **symmetrisch** um ihren Mittelwert $\mu$.  
  - Das bedeutet: $P(X \le \mu - a) = P(X \ge \mu + a)$ für alle $a$.  
  - Symmetrie erleichtert das Interpretieren von Wahrscheinlichkeiten und Extremwerten.  

- **Standardisierung**: Um Variablen vergleichbar zu machen oder Standardtabellen zu nutzen, transformiert man eine Zufallsvariable $X \sim N(\mu, \sigma^2)$ in die **Standardnormalverteilung** $Z \sim N(0,1)$:

$$
Z = \frac{X - \mu}{\sigma}
$$

- Vorteil:  
  - Daten verschiedener Skalen können direkt verglichen werden.  
  - Wahrscheinlichkeiten können aus Standardnormaltabellen abgelesen werden.  

**Beispiel:**  
Ein Test hat Mittelwert $\mu = 100$ und Standardabweichung $\sigma = 15$.  
- Ein Schüler hat $X = 130$ Punkte.  
- Standardisierte Punktzahl:  

$$
Z = \frac{130 - 100}{15} = 2
$$

- Interpretation: Der Schüler liegt **2 Standardabweichungen über dem Mittelwert**, entsprechend kann man $P(X \ge 130) = P(Z \ge 2) \approx 0.0228$ ablesen.  

---

### 6.4 Bedingte & marginale Verteilungen

- **Bedingte Verteilung:** Beschreibt die Verteilung einer Zufallsvariablen $X$, **gegeben** dass eine andere Zufallsvariable $Y=y$ ist:

$$
f_{X|Y}(x|y) = \frac{f_{X,Y}(x,y)}{f_Y(y)}
$$

- **Marginale Verteilung:** Gibt die Wahrscheinlichkeitsverteilung einer einzelnen Variablen $X$ unabhängig von anderen Variablen. Man integriert über alle Werte der anderen Variablen:

$$
f_X(x) = \int_{-\infty}^{\infty} f_{X,Y}(x,y) \, dy
$$

- **Anwendung:**  
  - Bei **multivariaten Normalverteilungen** z. B. $X$ und $Y$ korreliert.  
  - Man kann die **bedingte Verteilung von $X$ bei gegebenem $Y$** analysieren oder die **marginale Verteilung von $X$** unabhängig von $Y$ betrachten.  

**Beispiel:**  
- $X$ = Körpergröße, $Y$ = Gewicht in einer Population.  
- Gemeinsame Verteilung $f_{X,Y}(x,y)$ ist bivariant normal.  
- Bedingte Verteilung $f_{X|Y}(x|Y=70)$ = Verteilung der Körpergrößen von Personen mit 70 kg Gewicht.  
- Marginale Verteilung $f_X(x)$ = Verteilung der Körpergrößen über die gesamte Population, **unabhängig vom Gewicht**.  

- Wichtig:  
  - Bedingte Verteilungen werden z. B. in **Bayesianischer Statistik** und **multivariaten Modellen** genutzt.  
  - Marginale Verteilungen helfen, **Variablen einzeln zu analysieren**, bevor man Abhängigkeiten untersucht.

### 6.5 Mixtures of Gaussians (GMM)

- Eine **Mixture of Gaussians (GMM)** ist ein Modell, das eine **Kombination mehrerer Normalverteilungen** verwendet, um komplexere Datenverteilungen zu beschreiben:

$$
f_X(x) = \sum_{k=1}^{K} \pi_k \, \mathcal{N}(x \mid \mu_k, \sigma_k^2)
$$

- **Parameter:**
  - $\pi_k$: Gewicht der $k$-ten Komponente, mit $\sum_{k=1}^{K} \pi_k = 1$  
    → Gibt an, wie stark jede Komponente zur Gesamtverteilung beiträgt.
  - $\mu_k$: Mittelwert der $k$-ten Normalverteilung
  - $\sigma_k$: Standardabweichung der $k$-ten Normalverteilung

- **Einsatzbereiche:**
  - Modellierung von **mehrgipfligen Verteilungen** (z. B. Körpergröße in einer gemischten Population aus Männern und Frauen)
  - **Clustering**: GMM kann zur probabilistischen Gruppierung von Datenpunkten genutzt werden
  - **Dichteabschätzung**: Schätzung der Wahrscheinlichkeitsdichte komplexer Daten, wenn einfache Normalverteilung zu ungenau ist

- **Beispiel:**  
  Angenommen, wir messen die Einkommensverteilung in einer Stadt mit zwei klar unterscheidbaren Gruppen:
  - Gruppe 1: Niedriges Einkommen, $X_1 \sim N(\mu_1=2000, \sigma_1=500)$
  - Gruppe 2: Hohes Einkommen, $X_2 \sim N(\mu_2=6000, \sigma_2=1000)$
  - Gewichte: $\pi_1 = 0.7, \pi_2 = 0.3$

  Die kombinierte Verteilung:
  $$
  f_X(x) = 0.7 \, \mathcal{N}(x \mid 2000, 500^2) + 0.3 \, \mathcal{N}(x \mid 6000, 1000^2)
  $$

  → Die Gesamtverteilung ist **zweigipflig** und kann mit einer einzelnen Normalverteilung schlecht modelliert werden.

- **Visualisierung:**  
  - Jede Komponente kann separat geplottet werden, um zu zeigen, wie sie zusammen die Gesamtverteilung bilden.
  - Die GMM erlaubt eine **flexible Modellierung**, die auch asymmetrische oder multimodale Daten gut abbildet.

- **Tipp für Machine Learning:**  
  - GMMs werden z. B. in **EM-Algorithmus (Expectation-Maximization)** verwendet, um Parameter $\pi_k, \mu_k, \sigma_k$ aus Daten zu schätzen.
  - Gut geeignet für **unsupervised learning**, wenn Klassenlabels nicht bekannt sind.


### Zusammenfassung

- **PDF** gibt die Dichte an einem Punkt, **CDF** die kumulierte Wahrscheinlichkeit.  
- Symmetrische Verteilungen lassen sich **standardisieren**.  
- **Bedingte und marginale Verteilungen** ermöglichen Analyse mehrerer Variablen.  
- **Mixtures of Gaussians** erweitern Normalverteilungen für komplexe, mehrgipflige Daten.  


```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider
from scipy.special import erf

def plot_normal_with_area(mu=0.0, sigma=1.0, a=-1.0, b=1.0):
    if sigma <= 0:
        print("Die Standardabweichung σ muss > 0 sein.")
        return
    if a > b:
        print("Bitte sicherstellen, dass a <= b.")
        return

    x = np.linspace(mu - 4*sigma, mu + 4*sigma, 500)
    pdf = (1/(sigma * np.sqrt(2*np.pi))) * np.exp(-(x-mu)**2/(2*sigma**2))
    cdf = 0.5 * (1 + erf((x - mu)/(sigma*np.sqrt(2))))

    plt.figure(figsize=(10,5))

    # PDF
    plt.subplot(1,2,1)
    plt.plot(x, pdf, label='PDF', color='blue')
    plt.axvline(mu, color='red', linestyle='--', label=f'Erwartungswert μ={mu}')
    # Bereich a-b markieren
    mask = (x >= a) & (x <= b)
    plt.fill_between(x[mask], 0, pdf[mask], color='orange', alpha=0.5, label=f'P({a}≤X≤{b})')
    plt.title('Normalverteilung PDF mit markiertem Bereich')
    plt.xlabel('x')
    plt.ylabel('Dichte')
    plt.legend()
    plt.grid(True)

    # CDF
    plt.subplot(1,2,2)
    plt.plot(x, cdf, label='CDF', color='green')
    plt.title('Normalverteilung CDF')
    plt.xlabel('x')
    plt.ylabel('F(x)')
    plt.grid(True)
    plt.legend()

    plt.tight_layout()
    plt.show()

    # Wahrscheinlichkeit berechnen
    P_ab = 0.5 * (erf((b - mu)/(sigma*np.sqrt(2))) - erf((a - mu)/(sigma*np.sqrt(2))))
    print(f"Erwartungswert E[X] = {mu}")
    print(f"Varianz Var(X) = {sigma**2}")
    print(f"Wahrscheinlichkeit P({a} ≤ X ≤ {b}) = {P_ab:.4f}")

# Interaktives Widget
interact(plot_normal_with_area,
         mu=FloatSlider(value=0.0, min=-5.0, max=5.0, step=0.1, description='Mittelwert μ'),
         sigma=FloatSlider(value=1.0, min=0.1, max=5.0, step=0.1, description='Std. Abw. σ'),
         a=FloatSlider(value=-1.0, min=-10.0, max=10.0, step=0.1, description='a'),
         b=FloatSlider(value=1.0, min=-10.0, max=10.0, step=0.1, description='b'));
```

## Beispiel für eine Mixtur von Gaussverteilungen

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact

def plot_gmm(mu1=0, sigma1=1, w1=0.5,
             mu2=3, sigma2=1, w2=0.5,
             mu3=6, sigma3=1, w3=0.0):
    
    # Normiere die Gewichte
    weights = np.array([w1, w2, w3])
    weights = weights / np.sum(weights)
    
    x = np.linspace(-5, 10, 1000)
    
    # Einzelne Gauss-PDFs
    pdf1 = weights[0] * (1/(sigma1 * np.sqrt(2*np.pi))) * np.exp(-(x - mu1)**2 / (2*sigma1**2))
    pdf2 = weights[1] * (1/(sigma2 * np.sqrt(2*np.pi))) * np.exp(-(x - mu2)**2 / (2*sigma2**2))
    pdf3 = weights[2] * (1/(sigma3 * np.sqrt(2*np.pi))) * np.exp(-(x - mu3)**2 / (2*sigma3**2))
    
    pdf_total = pdf1 + pdf2 + pdf3
    
    plt.figure(figsize=(10,5))
    plt.plot(x, pdf_total, label='GMM Gesamt', color='black', linewidth=2)
    if weights[0]>0:
        plt.plot(x, pdf1, '--', label='Komponente 1')
    if weights[1]>0:
        plt.plot(x, pdf2, '--', label='Komponente 2')
    if weights[2]>0:
        plt.plot(x, pdf3, '--', label='Komponente 3')
    
    plt.title('Mixture of Gaussians (GMM)')
    plt.xlabel('x')
    plt.ylabel('Dichte')
    plt.legend()
    plt.grid(True)
    plt.show()

interact(plot_gmm,
         mu1=(-5,5,0.1), sigma1=(0.1,3,0.1), w1=(0.0,1.0,0.05),
         mu2=(-5,10,0.1), sigma2=(0.1,3,0.1), w2=(0.0,1.0,0.05),
         mu3=(-5,10,0.1), sigma3=(0.1,3,0.1), w3=(0.0,1.0,0.05));
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider, IntSlider

def plot_bivariate_gmm(mu1_x=0, mu1_y=0, sigma1=1,
                       mu2_x=3, mu2_y=3, sigma2=1,
                       weight1=0.5, x_cond=0):
    
    # Gewicht der zweiten Komponente
    weight2 = 1 - weight1
    
    # Gitter erstellen
    x = np.linspace(-5, 8, 200)
    y = np.linspace(-5, 8, 200)
    X, Y = np.meshgrid(x, y)
    
    # Dichte jeder Komponente
    def gaussian2d(X, Y, mu_x, mu_y, sigma):
        return (1/(2*np.pi*sigma**2)) * np.exp(-((X-mu_x)**2 + (Y-mu_y)**2)/(2*sigma**2))
    
    Z1 = weight1 * gaussian2d(X, Y, mu1_x, mu1_y, sigma1)
    Z2 = weight2 * gaussian2d(X, Y, mu2_x, mu2_y, sigma2)
    Z = Z1 + Z2
    
    # Marginale Dichten
    X_vals = x
    Y_vals = y
    marg_X = weight1 * (1/(np.sqrt(2*np.pi)*sigma1)) * np.exp(-(X_vals - mu1_x)**2/(2*sigma1**2)) + \
             weight2 * (1/(np.sqrt(2*np.pi)*sigma2)) * np.exp(-(X_vals - mu2_x)**2/(2*sigma2**2))
    marg_Y = weight1 * (1/(np.sqrt(2*np.pi)*sigma1)) * np.exp(-(Y_vals - mu1_y)**2/(2*sigma1**2)) + \
             weight2 * (1/(np.sqrt(2*np.pi)*sigma2)) * np.exp(-(Y_vals - mu2_y)**2/(2*sigma2**2))
    
    # Bedingte Verteilung Y|X=x_cond
    cond_Y = weight1 * (1/(np.sqrt(2*np.pi)*sigma1)) * np.exp(-(x_cond - mu1_x)**2/(2*sigma1**2)) * \
             (1/(np.sqrt(2*np.pi)*sigma1)) * np.exp(-(Y_vals - mu1_y)**2/(2*sigma1**2)) + \
             weight2 * (1/(np.sqrt(2*np.pi)*sigma2)) * np.exp(-(x_cond - mu2_x)**2/(2*sigma2**2)) * \
             (1/(np.sqrt(2*np.pi)*sigma2)) * np.exp(-(Y_vals - mu2_y)**2/(2*sigma2**2))
    cond_Y /= cond_Y.sum()  # Normalisierung
    
    # Plotten
    fig, axs = plt.subplots(2,2, figsize=(12,10))
    
    # 2D Heatmap
    im = axs[0,0].contourf(X, Y, Z, levels=50, cmap='viridis')
    axs[0,0].set_title("Bivariate GMM - Gesamtdichte")
    axs[0,0].set_xlabel("X")
    axs[0,0].set_ylabel("Y")
    fig.colorbar(im, ax=axs[0,0])
    
    # Marginale X
    axs[0,1].plot(X_vals, marg_X, label='Marginal X')
    axs[0,1].set_title("Marginale Dichte X")
    axs[0,1].legend()
    
    # Marginale Y
    axs[1,0].plot(Y_vals, marg_Y, label='Marginal Y', color='orange')
    axs[1,0].set_title("Marginale Dichte Y")
    axs[1,0].legend()
    
    # Bedingte Y|X=x_cond
    axs[1,1].plot(Y_vals, cond_Y, label=f'Bedingte Dichte Y|X={x_cond}', color='red')
    axs[1,1].set_title(f"Bedingte Verteilung Y|X={x_cond}")
    axs[1,1].legend()
    
    plt.tight_layout()
    plt.show()

# Interaktives Widget
interact(plot_bivariate_gmm,
         mu1_x=FloatSlider(value=0, min=-5, max=5, step=0.1, description='mu1_x'),
         mu1_y=FloatSlider(value=0, min=-5, max=5, step=0.1, description='mu1_y'),
         sigma1=FloatSlider(value=1, min=0.1, max=3, step=0.1, description='sigma1'),
         mu2_x=FloatSlider(value=3, min=-5, max=5, step=0.1, description='mu2_x'),
         mu2_y=FloatSlider(value=3, min=-5, max=5, step=0.1, description='mu2_y'),
         sigma2=FloatSlider(value=1, min=0.1, max=3, step=0.1, description='sigma2'),
         weight1=FloatSlider(value=0.5, min=0.0, max=1.0, step=0.05, description='weight1'),
         x_cond=FloatSlider(value=0, min=-5, max=8, step=0.1, description='X für Y|X'))
```
